### Άσκηση 1 - Πώς μπορούμε να υπολογίσουμε ένα ολοκλήρωμα στον υπολογιστη?

Ας κάνουμε μια άσκηση για ζέσταμα πριν μπούμε σε πιο ενδιαφέρουσες ασκήσεις παρακάτω... 

Η τιμή κάποιων απλών ολοκληρωμάτων μπορεί να υπολογιστεί με αρκετή ακρίβεια με χρήση της `Python`. Ας δούμε πρώτα τι λέει η θεωρία, και συγκεκριμένα ας κάνουμε μια βόλτα από τον Απειροστικό Λογισμό Ι και να θυμηθούμε το

<center> <strong> <font size=5>Ολοκλήρωμα Riemann </strong> </center>
<font size=4>
<br>
Ο Β. Riemann πρότεινε την προσέγγιση ενός ολοκληρώματος από τμηματικά σταθερές συναρτήσεις, των οποίων το αθροιστικό εμβαδό δίνει μια τιμή για το ολοκλήρωμα. Με μαθηματικά:

$$ E = \int_{a}^{b} f(t) dt = \lim_{\Delta t \rightarrow 0} \sum_{i=1}^{N} f(\bar{t}_i) \Delta t$$

Τι μας λέει η παραπάνω σχέση; Μας λέει ότι η τιμή ενός ορισμένου ολοκληρώματος μπορεί να βρεθεί αν κάνουμε τα εξής:

(i) Χωρίζουμε το διάστημα $[a,b]$ σε $N$ ίσα μέρη μήκους (διάρκειας) $\Delta t$ το καθένα. Αυτή η *διαμέριση* του διαστήματος λέγεται *ομοιόμορφη*.

(ii) Παίρνουμε ένα σημείο $\bar{t}_i$ σε κάθε ένα από αυτά τα διαστήματα. Για παράδειγμα, το μέσο του διαστήματος $ \Delta t $ ή όποιο άλλο θέλουμε.

(iii) Υπολογίζουμε την τιμή της συνάρτησης $f(t)$ στα σημεία $\bar{t}_i$ κάθε τμήματος της διαμέρισης.

(iv) Έχουμε τώρα $N$ τιμές της $f(t)$, δηλ. τις $f(\bar{t}_i)$, και πολλαπλασιαζόμενες μαζί με τα $N$ το πλήθος $\Delta t$, τις προσθέτουμε όλες μαζί για να πάρουμε το αποτέλεσμα.

Για να είναι ακριβές το αποτέλεσμα, πρέπει το $\Delta t$ να είναι όσο μικρότερο γίνεται, δηλ. ιδανικά να ισχύει $\Delta t \rightarrow 0$, ώστε να υπάρχει ισότητα μεταξύ ολοκληρώματος και αθροίσματος.

**Πρακτικά**, το παραπάνω μεταφράζεται στην προσέγγιση της επιφάνειας κάτω από τη συνάρτηση $f(t)$ από ορθογώνια παραλληλόγραμμα με "ύψος" $f(\bar{t}_i)$ και "πλάτος" $\Delta t$. Γνωρίζουμε από τη Γεωμετρία ότι κάθε παραλληλόγραμμο έχει επιφάνεια 

$$E_i = f(\bar{t}_i)\Delta t$$

και η προσέγγιση του ολοκληρώματος, υποθέτοντας ότι έχουμε $N$ τέτοια παραλληλόγραμμα, θα είναι

$$E \approx \sum_{i=1}^N E_i =\sum_{i=1}^N f(\bar{t}_i)\Delta t $$

Όταν το "πλάτος" $\Delta t$ κάθε παραλληλογράμμου γίνεται όλο και μικρότερο, τόσο καλύτερη η προσέγγιση της επιφάνειας (δηλ. του ολοκληρώματος που ζητάμε) από το άθροισμα των επιφανειών των παραλληλογράμμων.

Η διαδικασία φαίνεται στο παρακάτω σχήμα.

![riemann](./riemann.png)

I. Ας δούμε πως θα υλοιποιήσουμε αριθμητικά στη `Python` το ολοκλήρωμα Riemann

$$\int_{-\infty}^{+\infty}\left(-\frac{1}{3}\right)e^{-2(t-2)}u(t-2)dt = \int_2^{+\infty}\left(-\frac{1}{3}\right)e^{-2(t-2)}dt = -\frac{1}{6}$$

Έχουμε:

````python
# Ορίζουμε τον άξονα του χρόνου από t="2 + κάτι μικρό" (λόγω της βηματικής που ξεκινά από το t=2) ...
# ως μια μεγάλη τιμή που προσομοιώνει το άπειρο, π.χ. την τιμή 1000, τμηματοποιώντας τον ανά Dt = 0.001:
Dt = 0.001

# Άξονας του χρόνου:
t = np.arange(start=2+Dt, stop=1000, step=Dt)

# Η συνάρτηση x(t) ορίζεται ως:
x = (-1/3) * np.exp(-2*(t-2))

# Υπολογίζουμε με τον παρακάτω τρόπο το ολοκλήρωμα Riemann:
apotelesma = Dt * sum(x)

# Και το τυπώνουμε:
print(f'{apotelesma:.4f}')
````

Ας το επιβεβαιώσουμε:

````python
# Επιβεβαίωση:
print(f'{-1/6:.4f}')
````

II. Ας υπολογίσουμε τώρα το (διάσημο) ολοκλήρωμα

$$ \int_{0}^{\pi /2} \left(\frac{t}{\sin(t)}\right)^2 dt = \pi \ln (2) $$

Αρχικά πρέπει να προσέξουμε ότι η συνάρτηση μέσα στο ολοκλήρωμα δεν ορίζεται για $t=0$. Οπότε:

````python
# Ορίζουμε τον άξονα του χρόνου από "0 + κάτι_μικρό" ως π/2 για να εξασφαλίσουμε ότι ...
# δε μηδενίζεται ο παρονομαστής της συνάρτησής μας. Η τμηματοποίηση του άξονα είναι ...
# ξανά ανά Dt που τώρα το θέτουμε ίσο με 0.00001 ...
# Math notation: 1e-a = 10**(-a):
Dt = 1e-5 

# Άξονας του χρόνου:
t = np.arange(start=Dt, stop=np.pi/2, step=Dt)

# Ορίζουμε τη συνάρτηση t/sin(t):
f = (t/np.sin(t))

# Υψώνουμε στο τετράγωνο:
f_sq = f**2

# Υπολογίζουμε το ολοκλήρωμα Riemann όπως και πριν:
apotelesma = Dt * sum(f_sq)

# Και το τυπώνουμε:
print(f'{apotelesma:.4f}')
````

````python
# Επιβεβαίωση:
print(f'{np.pi*np.log(2):.4f}')
````

III. Με βάση τα παραπάνω, επιβεβαιώστε αριθμητικά στην `Python` τα ολοκληρώματα:

a. $$ \int_{0}^{1} \frac{\ln(t)}{1+t}dt = -\frac{\pi^2}{12} $$


b. $$ \int_{1}^{2} \frac{(t+2)(t-2)}{t^2} dt = -1 $$


c. $$ \int_{0}^{+\infty} \frac{t}{e^t - 1} dt = \frac{\pi^2}{6} $$

````python
# a.
Dt = 1e-5 
t1 = np.arange(start = Dt, stop = 1 + Dt, step = Dt)
f1 = np.log(t1) / (1 + t1)
apotelesma1 = Dt * sum(f1)
print(f"a. (Riemmann)   = {apotelesma1:.4f}")
print(f"a. (Επιβεβαίωση): -pi**2/12 = {-np.pi**2/12:.4f}")
````

````python
# b.
Dt = 1e-5
t2 = np.arange(start = 1, stop = 2 + Dt, step = Dt)
f2 = ((t2 + 2) * (t2 - 2)) / (t2 ** 2)
apotelesma2 = Dt * sum(f2)
print(f"b. (Riemmann) = {apotelesma2:.4f}")
print(f"b. (Επιβεβαίωση): {-1:.4f}")
````

````python
# c.
Dt = 1e-5
t3 = np.arange(start = Dt, stop = 999, step = Dt)
f3 = t3 / (np.exp(t3) - 1)
apotelesma3 = Dt * sum(f3)
print(f"c. (Riemmann)   = {apotelesma3:.4f}")
print(f"c. (Επιβεβαίωση): pi^2/6 = {np.pi**2/6:.4f}")
````

Αν και υπάρχει ξεχωριστή εξίσωση για τον υπολογισμό της ενέργειας ενός σήματος **διακριτού χρόνου** (και θα τη δούμε στην επόμενη άσκηση), όπως ένα .WAV αρχείο, θα χρησιμοποιήσουμε το ολοκλήρωμα Riemann για να υπολογίσουμε την ενέργεια, όπως είδατε στην προηγούμενη άσκηση (και θα δείτε και στο ακόλουθο toy example).

Το σήμα ομιλίας που έχουμε ηχογραφήσει και σας δίνουμε είναι *ψηφιακό* και όταν το διαβάζετε στην `Python`, το μετατρέπετε σε *διακριτού χρόνου*, αποθηκεύοντας τα δείγματά του σε ένα διάνυσμα. 

Μέχρι τότε όμως, ας καταλάβουμε λίγο τι συμβαίνει με ένα πιο ελεγχόμενο toy-παράδειγμα. Έστω ότι το σήμα ομιλίας διακριτού χρόνου που διαβάζετε είναι η ακολουθία από αριθμούς αποθηκευμένους στο διάνυσμα ```x``` όπως παρακάτω. Ας το δούμε.

````python
x = np.asarray([1, 0.5, 1.2, -1.7, 2.4, -0.3, -1.1, 0.6, -0.9])   # Τυχαίο σήμα - υποτίθεται σήμα ομιλίας.
print(f"Το σήμα ομιλίας διακριτού χρόνου είναι το x = {x}")       # Βλέπουμε τις τιμές του.
plt.stem(np.arange(0,9), x)                                       # Τυπώνουμε τις τιμές του.
plt.xlabel('Διακριτός χρόνος')                                    # Ομορφαίνουμε.
plt.ylabel('Πλάτος σήματος')                                      # Ομορφαίνουμε.
````

Οι τιμές που βλέπετε αποτελούν **δείγματα** του σήματος ομιλίας συνεχούς χρόνου που καταγράφηκε. Τα δείγματα αυτά λήφθηκαν ανά $1/F_s$ δευτερόλεπτα πραγματικού χρόνου, με $F_s$ να είναι η συχνότητα δειγματοληψίας, που σας δίνεται έτοιμη κάθε φορά που ανοίγετε ένα αρχείο ήχου στην `Python`. Αν, για παράδειγμα, η συχνότητα δειγματοληψίας είναι $16000$ Hz, τότε το παραπάνω σήμα $9$ δειγμάτων προέρχεται από ένα σήμα συνεχούς χρόνου διάρκειας 

$$(9-1)/16000 = 8/16000 = 500 \text{ μs} = 0.5 \text{ ms} = 0.0005 \text{ s}$$

Πραγματικά πολύ μικρής διάρκειας, αλλά είπαμε: toy-παράδειγμα. 😉 

Για να φτάσουμε στα $30$ ms που αναφέρει παραπάνω η άσκηση, χρειαζόμαστε $481$ δείγματα - αφού $(481-1)/16000 = 0.03$ s $= 30$ ms. Ας συνεχίσουμε όμως σε αυτό το toy example και ας υποθέσουμε όμως ότι θέλουμε να υπολογίσουμε την ενέργεια ανά $187.5$ μs, δηλ. ανά μόλις $N=3$ δείγματα. Τότε θα χωρίσουμε το σήμα σε $N-$άδες (τριάδες, εν προκειμένω), που ουσιαστικά αυτό αντιστοιχεί σε διαδοχικούς πολλαπλασιασμούς με μη επικαλυπτόμενους τετραγωνικούς παλμούς μοναδιαίου πλάτους και διάρκειας $187.5$ μs, ήτοι διάρκειας $N$ (τριών) δειγμάτων. 

Έτσι προκύπτουν (δείτε ξανά το διάνυσμα ```x``` παραπάνω) τρεις τριάδες, $(1, 0.5, 1.2), (−1.7, 2.4,−0.3), (−1.1, 0.6,−0.9)$, που καθεμιά είναι ένα παράθυρο σήματος, και θα υπολογίσουμε την ενέργειά καθεμιάς. Θα χρησιμοποιήσουμε μία λίστα (https://www.w3schools.com/python/python_lists.asp) της `Python`, στην οποία μπορούμε να *προσαρτήσουμε* (append) οποιαδήποτε μεταβλητή οποιουδήποτε μεγέθους, για να κρατάμε την ενέργεια καθενός παραθύρου.

````python
Fs = 16000   # Έστω μια συχνότητα δειγματοληψίας.
Dt = 1/Fs    # Περίοδος δειγματοληψίας (κάθε πόσα δευτερόλεπτα πήραμε ένα δείγμα από την ηχογράφησή μας).
N = 3        # Διάρκεια παραθύρου σε δείγματα : αυτό αντιστοιχεί σε 187.5 μικρο-δευτερόλεπτα πραγματικού χρόνου.

# Τμηματοποίηση σε διαδοχικές N-άδες και υπολογισμός ενέργειας τους:
x1 = x[0:N]     # από 0 ως Ν-1 ...
x2 = x[N:2*N]   # από Ν ως 2Ν-1 ...
x3 = x[2*N:3*N] # από 2Ν ως 3Ν-1 ...

# Άδεια λίστα για να μαζεύουμε τις ενέργειες των Ν-άδων:
E = []
E.append(Dt * np.sum(x1**2)) # Riemann άθροισμα για το 1ο τμήμα ...
E.append(Dt * np.sum(x2**2)) # Riemann άθροισμα για το 2ο τμήμα ...
E.append(Dt * np.sum(x3**2)) # Riemann άθροισμα για το 3ο τμήμα ...

print(f"E1 = {E[0]}\nE2 = {E[1]}\nE3 = {E[2]}")
# θα πρέπει να λάβετε:
# E1 = 0.000168125
# E2 = 0.000546249
# E3 = 0.00014875
````

Βλέπετε ότι κάθε τμήμα από το αρχικό σήμα έχει διαφορετική τιμή για την ενέργειά του. 

Ας επιστρέψουμε τώρα στο πλήρες, πραγματικό παράδειγμα, όπου θα υλοποιήσουμε πιο αυτοματοποιημένα τα παραπάνω, καθώς και την επιλογή της δραστηριότητας ομιλίας με βάση την ενέργεια κάθε τμήματος σήματος.

Ας γράψουμε κώδικα που διαβάζει ένα σήμα φωνής και μας υπολογίζει ένα διάνυσμα της ενέργειάς του σε κάποιες χρονικές στιγμές που αντιστοιχούν στα κέντρα των παραθύρων που αναλύουμε.

Η ιδέα είναι η εξής:

- Το σήμα ομιλίας θα αποθηκευτεί σε ένα ``numpy`` array.
- Θα το προσπελαύνουμε ανά μη επικαλυπτόμενα τμήματα, χρησιμοποιώντας array slicing, όπως στο παραπάνω toy example.
- Για κάθε τμήμα, θα υπολογίζουμε την ενέργειά του.
- Θα την αποθηκεύουμε σε ένα άλλο array, με τέτοιο τρόπο ώστε η $i-$οστή τιμή του array να αντιστοιχεί στο $i-$οστό τμήμα.
- Μόλις μαζέψουμε όλες τις ενέργειας, θα αποφασίσουμε ένα κατώφλι (threshold) με το οποίο θα συγκρίνουμε αυτές τις τιμές για να ταξινομούμε κάθε frame που αντιστοιχεί στις υπολογισμένες ενέργειες στην κατηγορία "speech" ή "non-speech".

Συμπληρώστε τον παρακάτω σκελετό.

````python
# Φόρτωση αρχείου ήχου στην Python (Fs είναι η συχνότητα δειγματοληψίας):
Fs, s = wav.read('speech.wav')

# Βήμα στο χρόνο (κάθε πότε πήραμε δείγμα από την πραγματική ηχογράφηση --> κάθε 1/Fs δευτερόλεπτα):
Dt = 1/Fs

# Κανονικοποίηση του σήματος σε πλάτη [-1,1] (τεχνική λεπτομέρεια, μην τη λάβετε υπ'όψη):
s = s/np.max(np.abs(s))

# θέλουμε παράθυρα 30ms στο χρόνο, σε πόσα δείγματα αντιστοιχούν? Αν 1 δευτ. αντιστοιχεί σε Fs δείγματα, πόσα δείγματα αντιστοιχούν σε 0.03 δευτ.? Χρησιμοποιήστε τη round() ...
T = round(0.03 * Fs)

# Συνολική διάρκεια σήματος:
L = len(s)

# Πόσα παράθυρα διάρκειας T δειγμάτων που βρηκαμε παραπάνω, μη επικαλυπτόμενα, χωρούν σε όλο το σήμα διάρκειας L δειγμάτων? Χρησιμοποιήστε τη round() ...
N = round(L / T)

# Δέσμευση μνήμης - αντίστοιχο της calloc στη C:
Energy = np.zeros(N)
Time = np.zeros(N)

# Επανάληψη για όλα τα παράθυρα που έχουμε:
for i in range(1, N):

    # Χρησιμοποιώντας το δείκτη i, τη διάρκεια T, και τον τελεστή int(), κόψτε κάθε φορά ένα τμήμα του σήματος s διάρκειας T:
    windowed_speech = s[int(i * T): int((i + 1) * T)]

    # Υπολογίστε την ενέργεια του τμήματος σήματος windowed_speech (όπως δείξαμε παραπάνω):
    Energy[i] = Dt * np.sum(windowed_speech ** 2)

    # Κέντρο του κάθε τμήματος (το δείγμα στη μέση του τμήματος, π.χ. ένα τμήμα διάρκειας 200 δειγμάτων έχει κέντρο το 100):
    Time[i] = round(i*T+T/2)
````

Ας επιλέξουμε κατώφλι απόφασης με βάση τα frames της σιωπής στην αρχή του αρχείου. Όπως είπαμε, τα πρώτα 2 δευτερόλεπτα της ηχογράφησης (δηλ. τα πρώτα $2\cdot F_s$ δείγματα) έχουν σιωπή. Συμπληρώστε κατάλληλα παρακάτω.

````python
# Κατώφλι απόφασης:
No_Silence_Frames = round(2*Fs/T) # το πληθος των frames που αντιστοιχούν στα πρώτα 2 δευτερόλεπτα ...
Threshold = Energy[:No_Silence_Frames].mean() # έστω ότι το κατώφλι είναι η μέση ενέργεια αυτών των πρώτων frames που αντιστοιχούν στα πρώτα 2 seconds ...

# Δέσμευση μνήμης για το διάνυσμα απόφασης (διάνυσμα με μηδενικά):
D = np.zeros(N)

# Διατρέχετε το διάνυσμα της ενέργειας που υπολογίσαμε ανά frame στον προηγούμενο κώδικα και αποφασίζετε:
# - 1 αν η ενέργεια του i-οστού frame είναι > από το κατώφλι ...
# - 0 αν η αντίστοιχη ενέργεια για το i-οστό frame είναι < κατώφλι ...
for i in range(N):
    if Energy[i] > Threshold: D[i] = 1
    else: D[i] = 0
````

Τώρα μπορούμε να "ενώσουμε" τις εκτιμήσεις μας και να απεικονίσουμε το αποτέλεσμα.

````python
plt.figure(figsize=(10,5))                                   # Γράφημα ...
t = np.arange(start=0, stop=L*Dt, step=Dt)                   # Άξονας χρόνου με βήμα Dt.
n = np.arange(start=0, stop=L, step=1)                       # Άξονας διακριτού χρόνου για κάθε δείγμα.
Din = np.interp(x=n, xp=Time, fp=D)                          # Γραμμική παρεμβολή.
plt.plot(t, Din, color='tab:blue', linewidth=2)              # Γράφημα απόφασης.
plt.plot(t, s/max(abs(s)), color='tab:red', linewidth=0.4)   # Γράφημα σήματος.
plt.title('Voice Activity Detector')                         # Ομορφαίνουμε.
plt.xlabel('Time (s)')                                       # -----||-----
plt.grid(linestyle=':')                                      # -----||-----
plt.show()                                                   # Εμφανίζουμε ...
````

Σας δίνεται η βάση δεδομένων ```Berlin-EMOdB```, μια αρκετά παλιά και γνωστή βάση δεδομένων συναισθηματικής ομιλίας στα Γερμανικά με 535 συνολικά αρχεία WAV. Η ομιλία είναι *acted*, δηλ. δεν είναι φυσικού συναισθήματος αλλά υποκριτικού (από ηθοποιούς). Πάμε να σετάρουμε τα βασικά paths.

Αν όλα πήγαν καλά στο παραπάνω κελί και βρήκατε τα $535$ αρχεία που πρέπει, ας δούμε τη βάση δεδομένων αυτή **περιγραφικά**: πόσα αρχεία WAV έχει, πώς είναι οργανωμένη, πόσους ομιλητές έχει, πόσα συναισθήματα, κλπ.

Εδώ πρέπει να γνωρίζετε ότι η βάση αυτή περιγράφει τα αρχεία της με το εξής στυλ (ενδεικτικό αρχείο): `03a01Fa.wav`  
Αυτή η ονοματοδοσία κωδικοποιεί τα πάντα για το αρχείο αυτό. Για παράδειγμα, το πρώτο '03' σημαίνει ότι μιλάει ο ομιλητής 03 και το γράμμα F δηλώνει το συναίσημα της "Χαράς".

Κάθε συναίσθημα κωδικοποιείται στο προτελευταίο γράμμα του WAV αρχείου ως εξής:

Σύμβολο : συναίσθημα (γερμανικό - ελληνικό όνομα)
- `W` : anger (Wut - Θυμός)
- `L` : boredom (Langeweile - Βαρεμάρα)
- `E` : disgust (Ekel - Αηδία)
- `A` : fear/anxiety (Angst - Φόβος)
- `F` : happiness (Freude - Χαρά)
- `T` : sadness (Trauer - Θλίψη)
- `N` : neutral (Ουδέτερο)

Ας ορίσουμε λοιπόν αυτήν την αντιστοίχιση και ας γράψουμε μια **κανονική έκφραση** (που ίσως ήδη ξέρετε ή αν όχι, θα μάθετε σε άλλα μαθήματα) για να εξάγουμε εύκολα αυτά τα στοιχεία από τη συμβολοσειρά του κάθε αρχείου WAV.

````python
# Μοτίβο για κάθε όνομα αρχείου (regular expression (RE) - κανονική έκφραση):
PRIMARY_RE = re.compile(r"^(?P<spk>\d{2})(?P<txt>[a-z])(?P<utt>\d{2})(?P<emo>[A-Z])(?P<int>[a-z])\.wav$", re.IGNORECASE)

# Αντιστοίχιση:
EMO_MAP = \
{
    "W": "Θυμός",
    "L": "Βαρεμάρα",
    "E": "Αηδία",
    "A": "Φόβος",
    "F": "Χαρά",
    "T": "Θλίψη",
    "N": "Ουδέτερο",
}
````

Πάμε να διαβάσουμε τα αρχεία της βάσης. 

Ξεκινάμε δημιουργώντας μια συνάρτηση που εντοπίζει τον ομιλητή και το συναίσθημα με βάση το μοτίβο της ονοματοδοσίας του αρχείου που ορίσαμε παραπάνω.

````python
def parse_emodb_filename(pathname):
    """ Επιστρέφει (speaker_id, emotion_code) """
    m = PRIMARY_RE.match(pathname.name) # έλεγξε το όνομα αρχείου με την regular expression ώστε να "ματσάρεις" τα στοιχεία ...
    spk = m.group("spk")                # πάρε τον κωδικό ομιλητή ...
    emo = m.group("emo").upper()        # πάρε τον κωδικό συναισθήματος ...
    return spk, emo                     # επέστρεψε ...
````

Τώρα πάμε να προσπελάσουμε τον κατάλογο της βάσης διαβάζοντας κάθε WAV αρχείο και να χρησιμοποιήσουμε την παραπάνω συνάρτηση για την εξαγωγή πληροφορίας/κατηγοριοποίηση των αρχείων.

````python
# ========== Προσπέλαση Βασης ==========

# Μια λίστα για να κρατάμε πληροφορίες:
rows = []

for wf in wav_files:                      # Για κάθε όνομα αρχείου στη λίστα wav_files που δημιουργήσαμε στην αρχή, στο πρώτο κελί της άσκησης...
    spk, emo = parse_emodb_filename(wf)   # ...βρες και πάρε τον κωδικό ομιλητή και τον κωδικό συναισθήματος από την ονοματοδοσία του αρχείου...
    sr, x = wav.read(str(wf))             # ...διάβασε και το WAV αρχείο και αποθήκευσέ το...
    dur = int(x.shape[0]) / float(sr)     # ...και υπολόγισε τη διάρκειά του σε δευτερόλεπτα!

    # Μια λίστα από dictionaries για να κρατάμε πληροφορίες για κάθε αρχείο WAV:
    rows.append(
    {
        "Αρχείο": wf.name,
        "Ομιλητής": spk,
        "Κωδικός συναισθήματος": emo,
        "Συναίσθημα": EMO_MAP.get(emo, "UNKNOWN"),
        "Fs": int(sr),
        "Πλήθος δειγμάτων": int(x.shape[0]),
        "Διάρκεια (s)": float(dur),
    }          )
````

Ας τυπώσουμε τι μάθαμε **γενικά** για τη βάση μας! 

````python
# ========== Στατιστικά ==========
speakers = [r["Ομιλητής"] for r in rows]
emos = [r["Κωδικός συναισθήματος"] for r in rows]
durations = np.array([r["Διάρκεια (s)"] for r in rows], dtype=float)
srs = np.array([r["Fs"] for r in rows], dtype=int)
print("\n=== Σύνοψη Βάσης Δεδομένων ===")
print("Πλήθος ομιλητών:", len(set(speakers)), sorted(set(speakers)))
print("Κωδικοί συναισθημάτων:", sorted(set(emos)))
print("Ονόματα συναισθημάτων:", [EMO_MAP.get(c, "UNKNOWN") for c in sorted(set(emos))])
print("Συχνότητα δειγματοληψίας (Hz):", sorted(set(srs.tolist())))
````

Πώς είναι κατανεμημένα τα WAV αρχεία ανά κλάση/συναίσθημα?

````python
from collections import Counter

# Μετράμε τα συναισθήματα και το πλήθος των WAV ανά συναίσθημα και το άθροισμά τους (535):
cc = Counter(emos)
N = sum(cc.values())

# Ας δούμε την κατανομή:
print("\nΠλήθος ηχογραφήσεων ανά κλάση:")
for c, k in sorted(cc.items(), key=lambda kv: (-kv[1], kv[0])):
    print(f"  {c} ({EMO_MAP.get(c,'UNKNOWN'):>8s}): {k:4d}  ({k/N:.2%})")
````

Ενδιαφέρον! Κάθε συναίσθημα δεν έχει το ίδιο πλήθος αρχείων! Υπάρχει μια ανισοκατανομή στη *συγκεκριμένη* βάση δεδομένων ανά συναίσθημα. Αυτό είναι κάτι που πρέπει να γνωρίζουμε.

Ας δούμε τη διάρκεια σε δευτερόλεπτα κάθε αρχείου ανά κλάση/συναίσθημα.

- **Εξαγωγή χαρακτηριστικών**: 

Παρ' όλο που υπάρχουν ΠΑΡΑ ΠΟΛΛΑ χαρακτηριστικά που μπορεί κανείς να εξάγει από την ομιλία, εμείς πάλι θα χρησιμοποιήσουμε την **ενέργεια** (μόνο!) κάθε σήματος, υπολογισμένη ανά παράθυρο (frame), όπως πριν. Αυτή τη φορά όμως θα το κάνουμε με μια μικρούλα παραλλαγή. Θα υπολογίσουμε την **Root Mean-Square Ενέργεια** η οποία δίνεται ως 

$$RMS_{energy} = \sqrt{\frac{1}{N}\sum_{i=1}^N x^2[n]}$$

για κάθε στοιχείο του διανύσματος $x$ που αντιπροσωπεύει ένα frame, και η οποία έχει κάμποσα πλεονεκτήματα σε σχέση με την "απλή" ενέργεια, ειδικά όταν θέλουμε να χτίσουμε αλγορίθμους που "μαθαίνουν" από αυτή.

$\underline{\textrm{Σημείωση:}}$ Στο συνεχή χρόνο που εξετάζουμε στο μάθημα, η ενέργεια αυτή θα γραφόταν ως

$$RMS_{energy} = \sqrt{\frac{1}{T}\int_{t_0}^{t_0+T}x^2(t)dt}$$

Υλοποιήστε παρακάτω μια συνάρτηση που υπολογίζει την RMS ενέργεια - όπως τη βλέπετε πιο πάνω, με το άθροισμα - για ένα frame ομιλίας (**εδώ πρέπει να συμπληρώσετε κώδικα**).

````python
# INSERT CODE:
def rms_energy(frame):
    N = len(frame)
    under_root = (np.sum(frame**2, axis=1) / frame.shape[1])        
    return np.sqrt(under_root + 1e-12)

# [Test] Πρέπει να πάρετε [1.73925271 6.40312424]: 
test_signal = np.asarray([[1.1, 2.2], [-1, -9]])
print(rms_energy(test_signal))
````

Νωρίτερα, στον ανιχνευτή ομιλίας με βάση την ενέργεια, χρησιμοποιήσαμε την ενέργεια ανά frame *χωρίς επικάλυψη* των γειτονικών frames. Όμως στην πραγματικότητα, κάποια επικάλυψη είναι πολλές φορές χρήσιμη για να παρακολουθούμε στενότερα την εξέλιξη της τιμής της ενέργειας προϊόντος του χρόνου. Γι' αυτό λοιπόν θα υιοθετήσουμε μια τέτοια προσέγγιση, όπου θα υπάρχει επικάλυψη κατά $50\%$ μεταξύ των γειτονικών frames. 

Σας δίνεται παρακάτω έτοιμη συνάρτηση που σας επιστρέφει ένα διδιάστατο πίνακα, όπου σε κάθε γραμμή του έχετε κι ένα frame διάρκειας $30$ ms, με επικάλυψη $15$ ms με το προηγούμενό του.

````python
def frame_signal(x, frame_len, hop):
    """
    Επιστρέφει έναν 2Δ πίνακα (n_frames, frame_len) που σε κάθε γραμμή του έχει ένα frame διάρκειας 
    frame_len δειγμάτων του συνολικού σήματος φωνής.
    Αν τα frames που θέλουμε δε "χωράνε" στο συνολικό σήμα, γίνεται ένα μικρό padding στο τέλος.
    """
    if len(x) < frame_len:
        x = np.pad(x, (0, frame_len - len(x)), mode="reflect")

    n_frames = 1 + int(np.ceil((len(x) - frame_len) / hop))
    total_len = (n_frames - 1) * hop + frame_len
    if total_len > len(x):
        x = np.pad(x, (0, total_len - len(x)), mode="reflect")

    shape = (n_frames, frame_len)
    strides = (x.strides[0] * hop, x.strides[0])
    
    return np.lib.stride_tricks.as_strided(x, shape=shape, strides=strides)
````

Αν λοιπόν καλέσουμε

```E = rms_energy(frame_signal(x, ...))```

θα πάρουμε ένα διάνυσμα ```Ε``` που θα έχει μια τιμή ενέργειας για κάθε frame ενός σήματος ομιλίας, δηλ. $N$ τιμές, όσες και τα frames ενός σήματος ομιλίας!

Καταλαβαίνετε όμως οτι

- Κάθε σήμα φωνής έχει τη δική του διάρκεια, δηλ. το πλήθος των frames του δεν είναι σταθερό.

- Το να χρησιμοποιήσουμε όλες τις τιμές ενέργειας για κάθε σήμα φωνής δε μας βοηθά. Πολλές τιμές, με μεγάλη συσχέτιση μεταξύ τους (πόση διαφορά να έχει η ενέργεια μεταξύ δυο γειτονικών frames ???).

Θα θέλαμε να **συνοψίσουμε στατιστικά** την "καμπύλη" ενέργειας για κάθε σήμα φωνής. Έτσι θα φτιάξουμε ένα νέο, *μικρότερο* διάνυσμα που περιγράφει στατιστικά το σήμα φωνής μέσω των χαρακτηριστικών της ενέργειάς του!

Πώς θα μπορούσαμε να περιγράψουμε στατιστικά τις τιμές της ενέργειας ενός σήματος? 

- Η **μέση τιμή** των τιμών της ενέργειας θα ήταν μια καλή ιδέα. Αρκεί? Μάλλον όχι.

- Αν προσθέταμε και την **τυπική απόκλιση** των τιμών της ενέργειας του σήματος? Κάπως καλύτερα... τι άλλο μπορούμε να βάλουμε? 

- Τη **μεσαία τιμή** (median) επίσης. 

Χμμμ... φτάνουν? Μπορεί, μπορεί και όχι. Ας προσθέσουμε μερικά ακόμα:

- **Ασυμμετρία (skewness)**: μετρά από ποιά μεριά βρίσκεται η "ουρά" της κατανομής των τιμών της ενέργειας. Θετική ασσυμετρία σημαίνει ότι η "ουρά" της κατανομής βρίσκεται προς τα δεξιά.

- **Κυρτότητα (kurtosis)**: περιγράφει ένα μέτρο που αφορά το σχήμα των "ουρών" της κατανομής σε σχέση με το συνολικό της σχήμα. Υψηλή κυρτότητα σημαίνει ότι οι "ουρές" της κατανομής είναι έντονες, υπάρχουν αρκετές τιμές στα άκρα της κατανομής.

Μέχρι τώρα έχουμε πέντε ($5$) στατιστικές τιμές ανά "καμπύλη" ενέργειας (δηλ. ανά WAV σήμα φωνής). Ούτε πολλές, ούτε λίγες. Υπάρχει ακόμα μια γνωστή τετράδα στατιστικών. Ας τη βάλουμε κι αυτή.

- **p10 (10ο εκατοστημόριο)**: Η τιμή κάτω από την οποία βρίσκεται το $10\%$ των τιμών της ενέργειας (δηλαδή "οι πιο χαμηλές τιμές").

- **q25 (25ο εκατοστημόριο)**: Η τιμή κάτω από την οποία βρίσκεται το $25\%$ των τιμών της ενέργειας.

- **q75 (75ο εκατοστημόριο)**: Η τιμή κάτω από την οποία βρίσκεται το $75\%$ των τιμών της ενέργειας.

- **p90 (90ο εκατοστημόριο)**: Η τιμή κάτω από την οποία βρίσκεται το $90\%$ των τιμών της ενέργειας (δηλαδή "οι πιο υψηλές τιμές").

Παρακάτω σας δίνεται μια συνάρτηση που υπολογίζει όλα αυτά τα στατιστικά από τις τιμές της ενέργειας ενός σήματος φωνής (**εδώ πρέπει να συμπληρώσετε κώδικα**).

````python
def summarize_stats(energy_vector):
    """
    Στατιστικά πάνω σε ένα διάνυσμα εισόδου energy_vector.
    Επιστρέφει, με τη σειρά: mean, std, skewness, kurtosis, median, p10, q25, q75, p90
    """
    ev = np.asarray(energy_vector, dtype=float).ravel()
    n = ev.size
    if n == 0:
        return np.zeros(10, dtype=float)

    mu = float(np.mean(ev))    # Κάντε χρήση της ev και της numpy function για τη μέση τιμή (mean) - κάντε cast σε float στο τέλος ...
    sd = float(np.std(ev))     # Κάντε χρήση της ev και της numpy function για την τυπική απόκλιση (standard deviation) - κάντε cast σε float στο τέλος ...
    med = float(np.median(ev)) # Κάντε χρήση της ev και της numpy function για το μεσαίο όρο (median) - κάντε cast σε float στο τέλος ...

    p10, q25, q75, p90 = np.percentile(ev, [10, 25, 75, 90]) # Τα εκατοστημόρια ...

    if sd > 1e-12:
        sk = float(skew(ev, bias=False))                     # Ασυμμετρία ...
        ku = float(kurtosis(ev, fisher=True, bias=False))    # Κύρτωση ...
    else:                                                    # Περίπτωση που δεν ορίζονται καλώς ...
        sk, ku = 0.0, 0.0

    return np.array([mu, sd, sk, ku, med, float(p10), float(q25), float(q75), float(p90)], dtype=float)

# Ας κωδικοποιήσουμε τα ονόματα για αυτά τα 9 χαρακτηριστικά:
FUNC_NAMES = ["mean", "std", "skew", "kurt", "median", "p10", "q25", "q75", "p90"]
feature_names = [f"E_{s}" for s in FUNC_NAMES]
print("n_features =", len(feature_names), feature_names)
````

Ας φτιάξουμε μια συνάρτηση ```extract_features``` που να δέχεται ένα path για ένα σήμα φωνής και να εξάγει αρχικά την ενέργεια, frame-by-frame, και στη συνέχεια τα παραπάνω $9$ χαρακτηριστικά! (**Εδώ πρέπει να συμπληρώσετε κώδικα**).

````python
def extract_features(wav_path, frame_ms=30.0, hop_ms=15.0):
    """
    Εξάγει την RMS ενέργεια από ένα σήμα φωνής και μετά τα 
    απαιτούμενα 9 χαρακτηριστικά από την ενέργεια αυτή.
    Επιστρέφει τα 9 στατιστικά.
    """
    sr, x = wav.read(wav_path)  # Διαβάζουμε το σήμα φωνής!
    x = np.asarray(x)           # Safety :)

    # Κανονικοποίηση:
    info = np.iinfo(x.dtype)
    denom = float(max(abs(info.min), abs(info.max)))
    x = x.astype(np.float32) / denom
    x = np.clip(x, -1.0, 1.0)
    
    # Παράμετροι για τη frame-by-frame ανάλυση:
    frame_len = int(round(sr * frame_ms / 1000.0))         # Μετατρέπουμε τα 30 ms σε δείγματα: 480 ...
    hop = int(round(sr * hop_ms / 1000.0))                 # Μετατρέπουμε τα 15 ms σε δείγματα: 240 ...
    frames = frame_signal(x, frame_len=frame_len, hop=hop) # Καλούμε την frame_signal για την τμηματοποίηση του σήματος.
    
    e = rms_energy(frames)       # Υπολογίζουμε την ενέργεια για όλα τα frames καλώντας κατάλληλη συνάρτηση ...
    e_stats = summarize_stats(e) # Επιστρέφουμε τα στατιστικά για την "καμπύλη" ενέργειας καλώντας κατάλληλη συνάρτηση ...

    return e_stats

# Testing:
f0 = extract_features(kept_files[0])
print(f0) 
# Πρέπει να πάρετε: [4.50768153e-02 5.79770227e-02 2.05577052e+00 5.07799827e+00 2.35119592e-02 9.99999997e-07 1.72353094e-03 6.95751421e-02 1.08393940e-01]
````

Απλά τρέξτε το παρακάτω κελί, μη δώσετε σημασία στις λεπτομέρειες. Η ουσία είναι ότι η συνάρτηση αυτή δημιουργεί δυο **δείκτες** στο αρχικό σύνολο των δεδομένων μας, που σέβονται τις δυο παραπάνω προϋποθέσεις.

````python
# ------------------------------------------------------------------------------------------------------------- #
# [SOS] New python cell: intalling the package because the below cell appeared as an error and was crashing ...
# %pip install scikit-learn
# ------------------------------------------------------------------------------------------------------------- #
````

````python
from sklearn.model_selection import StratifiedGroupKFold

# Χωρίζουμε τα δυο σετ, training & test, με βάση τους speakers και την αρχική κατανομή συναισθημάτων της βάσης:
def stratified_group_split(labels, speakers, seed=1337):
    n = len(labels)
    sgkf = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=seed)
    for tr, te in sgkf.split(np.zeros(n), labels, speakers):
        return np.asarray(tr), np.asarray(te)
````

Η συνάρτηση αυτή επιστρέφει δυο πίνακες: 

1. Έναν που περιέχει τους **δείκτες** στους πίνακες ```data```, ```label```, ```speakers``` που αντιστοιχούν στο training set. Αυτό αποτελεί περίπου το $80\%$ της επιλεγμένης βάσης.

2. Έναν που περιέχει τους **δείκτες** στους πίνακες ```data```, ```label```, ```speakers``` που αντιστοιχούν στο test set. Αυτό αποτελεί περίπου το $20\%$ της επιλεγμένης βάσης.

Ας τη δοκιμάσουμε!

````python
train_idx, test_idx = stratified_group_split(label, speakers, seed=RNG_SEED)

print(f"Μέγεθος training set: {len(train_idx)}, Μέγεθος test set: {len(test_idx)}\n")
print("Πλήθος ομιλητών στο training set:", sorted(set(speakers[train_idx])))
print("Πλήθος ομιλητών στο test set: ", sorted(set(speakers[test_idx])))

# Ας δούμε το ποσοστό αναλογίας κάθε συναισθήματος σε κάθε σετ (train & test):
def show_props(idx, name):
    labs = sorted(set(label))
    props = {lab: float(np.mean(label[idx] == lab)) for lab in labs}
    print(name, props)

# Ένα cross-check:
show_props(np.arange(len(label)), "\nΣυνολική βάση:")
show_props(train_idx, "Training  ")
show_props(test_idx,  "Test   ")
````

Βλέπετε ότι οι αναλογίες ανά συναίσθημα είναι κοντά στην αρχική αναλογία της βάσης, δεδομένης και της κατανομής των ομιλητών ανά συναίσθημα:

Φτάσαμε στο επιθυμητό σημείο. Τώρα πρέπει απλά να επιλέξουμε ένα μοντέλο μηχανικής μάθησης για να το εκπαιδεύσουμε με τα χαρακτηριστικά που φτιάξαμε: τα $9$ στατιστικά κάθε πρότασης, μαζί με τον κωδικό συναισθήματος, ώστε να μάθει την αντιστοιχία 

<p style="text-align:center;">
τα 9 χαρακτηριστικά της i-οστής πρότασης αντιστοιχούν στο j-οστό συναίσθημα
</p>

Από τα πραγματικά ΠΑΡΑ ΠΟΛΛΑ μοντέλα, θα επιλέξουμε ένα πολύ απλό. Το μοντέλο της **logistic regression για περισσότερες από δυο κλάσεις (multinomial logistic regression)**.

Με *πολύ απλά λόγια*, η multinomial logistic regression είναι ένας απλός ταξινομητής για πολλές κλάσεις: για κάθε κλάση $k$ υπολογίζει ένα "σκορ" ως γραμμικό συνδυασμό των χαρακτηριστικών $x$ (των στατιστικών ενέργειας, στη δική μας περίπτωση):

$$s_k = w_k^\top x + b_k$$

Μετά μετατρέπει όλα αυτά τα σκορ σε πιθανότητες που αθροίζουν στο 1 με τη συνάρτηση softmax: 

$$p_k = \frac{e^{s_k}}{\sum_j e^{s_j}}$$

Στην εκπαίδευση, το μοντέλο αλλάζει τα $w_k, b_k$ ώστε να δίνει όσο γίνεται μεγαλύτερη πιθανότητα στη σωστή κλάση (ελαχιστοποιώντας κάποια συνάρτηση κόστους). Στην πρόβλεψη, απλώς διαλέγει την κλάση με τη μεγαλύτερη πιθανότητα.

Όλα αυτά θα τα μάθετε με λεπτομέρεια σε μαθήματα Μηχανικής Μάθησης, και δεν αποτελεί βασικό στοιχείο αυτού του εργαστηρίου. 

Η βιβλιοθήκη ```scikit-learn``` έχει τα πάντα που θα χρειαστούμε. Απλά τρέξτε τον παρακάτω κώδικα που εκτελεί την εκπαίδευση και τον έλεγχο.

````python
# ------------------------------------------------------------------------------------------------------------- #
# [SOS] New python cell: please note that i commented out the line:
# > multi_class="multinomial",
# ... from the code of the sell below because an error was occurring and everything was crashing ...
# ------------------------------------------------------------------------------------------------------------- #
````

````python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

# Ορίζουμε το pipeline της εκπαίδευσης - don't worry about it:
clf = Pipeline([
    ("scaler", StandardScaler()),
    ("lr", LogisticRegression(
        max_iter=20000,
        solver="lbfgs",
        # multi_class="multinomial",
        random_state=RNG_SEED,
        class_weight="balanced",
        penalty="l2"
    ))])

# Επιλέγουμε το training και το test set με βάση τους δείκτες που πήραμε από την StratifiedGroupKFold νωρίτερα:
X_train, y_train = data[train_idx], label[train_idx]
X_test, y_test = data[test_idx], label[test_idx]

# Εκπαιδεύουμε τον αλγόριθμο:
clf.fit(X_train, y_train)

# Του ζητάμε να προβλέψει τα συναισθήματα στο test set!
y_pred = clf.predict(X_test)
````

Στο διάνυσμα ```y_pred``` υπάρχουν τα labels των συναισθημάτων που πρόβλεψε το μοντέλο για τις προτάσεις που "επιλέξαμε" να βρίσκονται στο test set!

````python
print("Προβλέψεις:", y_pred)   # Προβλέψεις.
print("Ground truth:", y_test) # Πραγματικά συναισθήματα (labels).
````

Αν θελήσουμε να δημιουργήσουμε στην `Python` ένα ημιτονοειδές σήμα συχνότητας $f_0=400$ Hz (νότα ΛΑ), διάρκειας $0.5$ sec (δηλ. $500$ ms), με μηδενική φάση μετατόπισης $\phi$ και μοναδιαίου πλάτους $A=1$, με συχνότητα δειγματοληψίας $f_s=16000$ Hz (που αντιστοιχεί σε καλής ποιότητας ηχογράφηση) θα γράψουμε τις ακόλουθες εντολές:

Στην άσκηση αυτή θα συνθέσουμε τμήμα από το πολύ γνωστό κομμάτι του Beethoven, Fur Elise (https://www.youtube.com/watch?v=q9bU12gXUyM), από νότες πλήκτρων πιάνου.

(α') Με χρήση της βιβλιοθήκης `Pandas` μπορούμε να διαβάσουμε ένα CSV αρχείο `FurEliseData.csv` που μας δίνεται

````python
df = pd.read_csv('FurEliseData.csv')
````

ως ένα `DataFrame`, και να το τυπώσουμε.

Όπως βλέπετε, υπάρχουν $4$ στήλες, `lefthand, righthand, ldur, rhdur` οι οποίες ανταποκρίνονται σε αριθμούς πλήκτρου πιάνου που παίζονται από το αριστερό χέρι (`lefthand`), με τις αντίστοιχες διάρκειές τους σε ms (`lhdur`), και σε αριθμούς πλήκτρου πιάνου που παίζονται με το δεξί χέρι (`righthand`), με τις αντίστοιχες επίσης διάρκειές τους σε ms (`rhdur`). Τα μηδενικά αντιστοιχούν σε παύσεις (δεν παίζουμε κάποιο πλήκτρο), αντίστοιχης διάρκειας.

Αυτούς μπορούμε να τους εισάγουμε σε ξεχωριστούς πίνακες της Python ως εξής:

````python
lefthand = df["lefthand"].values
righthand = df["righthand"].values
lhdur = df["lhdur"].values
rhdur = df["rhdur"].values

print(f"Τα πλήκτρα πιάνου του αριστερού χεριού είναι τα: \n {lefthand}")
print(f"Οι διάρκειες σε ms κάθε πλήκτρου του αριστερού χεριού είναι: \n {lhdur} ms")
print(f"Τα πλήκτρα πιάνου του δεξιού χεριού είναι τα: \n {righthand}")
print(f"Οι διάρκειες σε ms κάθε πλήκτρου του δεξιού χεριού είναι: \n {rhdur} ms")
````

(β') Μετατρέψτε από κάθε χέρι, κάθε αριθμό πλήκτρου πιάνου $n$ στην αντίστοιχη συνχότητα με τη σχέση
<br>
$$\Large f(n) = f_\mathrm{LA} \cdot 2^{\frac{n-49}{12}} = 440 \cdot 2^{\frac{n-49}{12}} $$
Στην `Python` αυτό γίνεται εύκολα ως

````python
# Μετατροπή πλήκτρου πιάνου σε νότα για το αριστερό χέρι:
freq_lh = 440 * (2 ** ((lefthand - 49) / 12))

# Μετατροπή πλήκτρου πιάνου σε νότα για το δεξί χέρι (συμπληρώστε) ?
freq_rh = 440 * (2 ** ((righthand - 49) / 12))
````

με αποτέλεσμα έναν πίνακα - διάνυσμα `freq` με τις αντίστοιχες συχνότητες. Οι παύσεις αναπαρίστανται στους πίνακες με τα πλήκτρα ως $(0)$. Θα μετατραπούν και αυτές - ενώ δεν πρέπει - όμως αυτό δε θα μας επηρεάσει, καθώς τις παύσεις θα τις χειριστούμε ξεχωριστά όπως θα δείτε παρακάτω.

(γ') Οι πίνακες `lhdur, rhdur` περιέχουν τη διάρκεια κάθε νότας σε ms για το αριστερό και το δεξί χέρι, αντίστοιχα. 

Ακολουθώντας τον παρακάτω κώδικα, δημιουργήστε τα αντίστοιχα ημίτονα με τη σωστή διάρκεια. Για τις παύσεις, θα δημιουργήσουμε ένα μηδενικό διάνυσμα με την ανάλογη διάρκεια.

````python
# Πόσα πλήκτρα έχουμε στο αριστερό χέρι ?
N = len(lefthand)

# Συχνότητα δειγματοληψίας:
fs = 16000

# Μια άδεια λίστα για να μαζεύουμε τα ημίτονα που θα φτιάχνουμε:
signal_left = []

# Ξεκινάμε:
for i in range(1, N):
    
    # Ο άξονας χρόνου για κάθε πλήκτρο:
    t = np.arange(start=0, stop=lhdur[i]/1000, step=1/fs)
    
    if lefthand[i] == 0:
        # Αν έχουμε παύση, δημιουργύμε ένα μηδενικό διάνυσμα διάρκειας lhdur[i] ms:
        signal_left = np.append(signal_left, np.zeros(len(t)))
    else:
        # Φτιάξτε το αντίστοιχο ημίτονο με χρήση του διανύσματος freq_lh (χρησιμοποιήστε την np.sin() και το παράδειγμα στην αρχή της άσκησης) ...
        new_sin = np.sin(2 * np.pi * freq_lh[i] * t)
        # "Μαζέψτε" όλα τα ημίτονα στη λίστα signal_left:
        signal_left = np.append(signal_left, new_sin)
````

(δ') Η μεταβλητή `signal_left` εδώ δηλώνεται ως μία λίστα (https://www.w3schools.com/python/python_lists.asp) της `Python`, στην οποία μπορούμε να *προσαρτήσουμε* (append) οποιαδήποτε μεταβλητή οποιουδήποτε μεγέθους. Αυτό είναι βολικό γιατί οι νότες δεν έχουν όλες την ίδια διάρκεια, και ένας πίνακας $N\times L$ θα ήταν δύσχρηστος μιας και το $L$ αλλάζει δυναμικά.

(ε') Επαναλάβετε τη διαδικασία για το δεξί χέρι, δηλ. τις μεταβλητές `righthand, rhdur` και αποθηκεύστε τα αποτελέσματά σας σε ξεχωριστές μεταβλητές. Παρατηρήστε ότι το αριστερό χέρι παίζει λιγότερες νότες από το δεξί.

````python
# Πόσα πλήκτρα έχουμε στο δεξί χέρι ?
N = len(righthand)

# Συχνότητα δειγματοληψίας:
fs = 16000

# Μια άδεια λίστα για να μαζεύουμε τα ημίτονα που θα φτιάχνουμε:
signal_right = []

# Ξεκινάμε:
for i in range(1, N):
    
    # Ο άξονας χρόνου για κάθε πλήκτρο:
    t = np.arange(start=0, stop=rhdur[i]/1000, step=1/fs)
    
    if righthand[i] == 0:
        # Αν έχουμε παύση, δημιουργύμε ένα μηδενικό διάνυσμα διάρκειας rhdur[i] ms:
        signal_right = np.append(signal_right, np.zeros(len(t)))
    else:
        # Φτιάξτε το αντίστοιχο ημίτονο με χρήση του διανύσματος freq_rh  (χρησιμοποιήστε την np.sin() και το παράδειγμα στην αρχή της άσκησης) ...
        new_sin = np.sin(2 * np.pi * freq_rh[i] * t)
        # "Μαζέψτε" όλα τα ημίτονα στη λίστα signal_right:
        signal_right = np.append(signal_right, new_sin)
````

(ζ) Δημιουργήστε το συνολικό σήμα, αθροίζοντας τα δύο σήματα ως

`sig = signal_left + signal_right`

````python
sig = signal_left + signal_right # Διαβάστε παρακάτω αν σας βγάλει error!
````

Αν τα έχετε κάνει όλα σωστά, η `Python` θα σας επιστρέψει σφάλμα, γιατί το ένα σήμα βγήκε λίγο μικρότερο από το άλλο, οπότε δε γίνεται να προστεθούν (να θυμάστε ότι οι πράξεις μεταξύ πινάκων πρέπει να τηρούν τους κανόνες διάστασης των πινάκων). Χρησιμοποιήστε την εντολή `len` και μετρήστε το μήκος του κάθε σήματος. Πόση είναι η διαφορά τους σε πλήθος τιμών;

````python
len_left = len(signal_left)
print(f"To μήκος σήματος αριστερού χεριού είναι {len_left} δείγματα.")

len_right = len(signal_right)
print(f"To μήκος σήματος δεξιού χεριού είναι {len_left} δείγματα.")

D = len_right - len_left
print(f"Το σήμα δεξιού χεριού είναι μεγαλύτερο κατά {D} δείγματα.")
````

(η') Για μια γρήγορη διόρθωση στο παραπάνω πρόβλημα, χρησιμοποιήστε τις εντολές

````python
D = len(signal_right) - len(signal_left)
signal_left = np.append(signal_left, np.zeros(D))
sig = signal_left + signal_right
````

όπου απλά συμπληρώνουμε με μηδενικά το μικρότερο διάνυσμα για να γίνει ίσου "μεγέθους" με το μεγαλύτερο.

Ο τερματισμός ενός ημιτόνου σε τυχαίο πλάτος και η αρχή του επομένου από μηδενικό πλάτος δημιουργεί μια απότομη αλλαγή, μια ασυνέχεια/αλλαγή φάσης, η οποία ακούγεται σαν click. Σε κάθε παράδειγμα του σχήματος βλέπετε δυο ημίτονα διαφορετικής συχνότητας σαν αυτά που δημιουργήσατε στην προηγούμενη άσκηση, με το ένα να ακολουθεί το άλλο. Αν οι διάρκεια του πρώτου ημιτόνου είναι κατάλληλα επιλεγμένη, τότε η συνένωση της δεύτερης νότας γίνεται ομαλά, όπως στο Σχήμα (α). Όμοια και στο Σχήμα (β), όπου η συνένωση είναι ακόμα πιο ομαλή και μια τέτοια περίπτωση είναι ιδανική. Και στις δυο περιπτώσεις τα clicks είναι ανεπαίσθητα ή ανύπαρκτα. 

Όμως αυτό που στην πραγματικότητα κάνατε στην προηγούμενη άσκηση είναι το Σχήμα (γ), καθώς η διάρκεια της κάθε νότας δεν τερμάτιζε απαραίτητα σε σημείο τέτοιο ώστε η ένωση της με το ημίτονο της επόμενης νότας να σχηματίζει "συνέχεια" στο σήμα. Τις περισσότερες - αν όχι όλες - φορές, ένα ημίτονο που δημιουργούσατε τερμάτιζε σε στιγμή που το πλάτος του δεν ήταν μηδενικό, και αναγκαστικά δημιουργούνταν μια ασυνέχεια στο ηχητικό σήμα λόγω της προσθήκης της επόμενης νότας, που ξεκινούσε από το μηδέν (αφού ήταν συναρτήσεις $\sin(\cdot)$, αλλά το ίδιο ϑα συνέβαινε αν χρησιμοποιούσαμε $\cos(\cdot))$.

Μια εύκολη λύση αυτού του προβλήματος είναι ο πολλαπλασιασμός του ημιτόνου που μοντελοποιεί κάθε νότα με ένα σήμα παραθύρου, ίδιας διάρκειας με το ημίτονο που παριστάνει κάθε νότα. Γνωρίζουμε ήδη δυο τέτοια παράθυρα: τον τετραγωνικό παλμό και τον τριγωνικό παλμό. Ουσιαστικά, ο τετραγωνικός παλμός χρησιμοποιείται έμμεσα κάθε ϕορά που δημιουργείτε ένα ημίτονο συγκεκριμένης διάρκειας. Όμως μπορείτε να χρησιμοποιήσετε άλλα παράθυρα, τέτοια ώστε να αναγκάζουν το ημιτονοειδές σήμα να κατεβαίνει ομαλά στο μηδέν στα άκρα του, όπως στο Σχήμα παρακάτω, όπου χρησιμοποιούμε το γνωστό παράθυρο Hanning, που έχει κωδωνοειδή μορφή. 

![win](./windowing.png)

Με αυτό τον τρόπο τα clicks μειώνονται σημαντικά, ενώ δημιουργούνται και κάποια "εξωτικά" εφέ στον ήχο που παράγεται.

Η βιβλιοθήκη `NumPy` της `Python` έχει έτοιμες συναρτήσεις παραθύρων για να χρησιμοποιήσετε. Όλες καλούνται με τον ίδιο τρόπο:

`window = np.window_type(duration)`

με το window_type να είναι ένα από τα εξής:

`hamming, hanning, bartlett, blackman`

και τη μεταβλητή `duration` να είναι η διάρκεια του παραθύρου. 

Εκτελέστε την παρακάτω συνάρτηση που αναλαμβάνει να πολλαπλασιάζει τα ημίτονα με το εκάστοτε παράθυρο.

````python
def piano_simulator(lhand, rhand, ldur, rdur, sr, window):
    """ Η συνάρτηση αυτή υλοποιεί την προηγούμενη άσκηση 
    και πολλαπλασιάζει κάθε ημίτονο με ένα γνωστό παράθυρο 
    για να μειώσει τα clicks μεταξύ γειτονικών ημιτόνων...
    Η μεταβλητή sr είναι η συχνότητα δειγματοληψίας. """

    if window == "hanning":
        win = np.hanning
    elif window == "hamming":
        win = np.hamming
    elif window == "bartlett":
        win = np.bartlett
    elif window == "blackman":
        win = np.blackman
    else:
        print(f"\"{window}\" was given as the window; please provide one of the following ones: \"hanning\", \"hamming\", \"bartlett\", \"blackman\".")
        return 0

    # Το αριστερό χέρι πρώτα:
    signal_left = []

    for i in range(1, len(lhand)):
        t = np.arange(start=0, stop=ldur[i]/1000, step=1/sr)
        if lhand[i] == 0:
            signal_left = np.append(signal_left, np.zeros(len(t)))
        else:
            freq = 440 * (2 ** ((lhand[i] - 49) / 12))
            # Η παραθυροποίηση γίνεται εδώ:
            signal_left = np.append(signal_left, np.sin(2*np.pi*freq*t) * win(len(t)))

    # Το δεξί χέρι μετά:
    signal_right = []

    for i in range(1, len(rhand)):
        t = np.arange(start=0, stop=rdur[i]/1000, step=1/sr)
        if rhand[i] == 0:
            signal_right = np.append(signal_right, np.zeros(len(t)))
        else:
            freq = 440 * (2 ** ((rhand[i] - 49) / 12))
            # κι εδώ:
            signal_right = np.append(signal_right, np.sin(2*np.pi*freq*t) * win(len(t)))

    # Χειριζόμαστε τη διαφορά στις διάρκειες όπως στην προηγούμενη άσκηση ...
    d = len(signal_right) - len(signal_left)
    signal_left = np.append(signal_left, np.zeros(d))

    return signal_left + signal_right
````

Ας δούμε τι παράγει λοιπόν η παραπάνω συνάρτηση για καθένα από τα παράθυρα που εφαρμόζουμε.

````python
# Συχνότητα δειγματοληψίας:
fs = 16000

# Φορτώνουμε ξανά τις μεταβλητές μας:
df = pd.read_csv('FurEliseData.csv')
lefthand = df["lefthand"].values
righthand = df["righthand"].values
lhdur = df["lhdur"].values
rhdur = df["rhdur"].values

# Καλούμε τη συνάρτηση για τα 4 γνωστά παράθυρα:
hann_out  = piano_simulator(lefthand, righthand, lhdur, rhdur, fs, "hanning" )
hamm_out  = piano_simulator(lefthand, righthand, lhdur, rhdur, fs, "hamming" )
bart_out  = piano_simulator(lefthand, righthand, lhdur, rhdur, fs, "bartlett")
black_out = piano_simulator(lefthand, righthand, lhdur, rhdur, fs, "blackman")
````
