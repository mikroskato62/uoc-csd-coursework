Στην προηγούμενη σειρά ασκήσεων είδαμε πως μπορούμε να υπολογίσουμε πράγματα γράφοντας εντολές η μια μετά την άλλη. Μπορούμε να μείνουμε σε αυτό το σχήμα προγραμματισμού αλλά θα ήταν μια καλύτερη ιδέα να δούμε πώς μπορούμε να γράψουμε συναρτήσεις στην Python. 

Οι συναρτήσεις είναι τμήμα κώδικα που δέχεται κάποια (ή και κανένα) ορίσματα και επιστρέφει κάποιο/α αποτέλεσμα/τα. Σκοπός της συγγραφής μιας συνάρτησης είναι η επαναχρησιμοποίησή της σε διάφορα μέρη ενός μεγαλύτερου προγράμματος. Η συνάρτηση δεν είναι απλά ένα σύνολο από εντολές, "πακεταρισμένες" κάτω από έναν ορισμό συνάρτησης αλλά συχνά επιτελούν ένα συγκεκριμένο ρόλο. 

Για παράδειγμα, μια συνάρτηση που λύνει ένα σύστημα $N$ x $N$ εξισώσεων έχει νόημα, ενώ μια συνάρτηση που επιστρέφει το τετράγωνο ενός αριθμού δεν έχει τόσο νόημα (είναι μια πολύ απλή πράξη που δε χρειάζεται να "κλειστεί" σε μια συνάρτηση, παρ' όλο που ίσως το τετράγωνο ενός αριθμού χρειάζεται να υπολογιστεί πολύ συχνά σε ένα μεγάλο πρόγραμμα).

Ας δούμε λοιπόν πως γράφουμε μια συνάρτηση η οποία υπολογίζει το παραγοντικό ενός ακέραιου αριθμού που της δίνουμε. Το παραγοντικό ενός ακέραιου αριθμού $n$ ορίζεται ως 

$$n! = 1\cdot 2 \cdot 3 \cdot \cdots \cdot (n-1) \cdot n$$

```python
def factorial(n):
    r = 1
    for i in range(1, n + 1):
        r = r * i
    return r
```

Η συνάρτηση $\textrm{range(a,b)}$ επιστρέφει μια ακολουθία από αριθμούς που ξεκινά από το $\textrm{a}$ και σταματά στο $\textrm{b-1}$ (ΟΧΙ στο $\textrm{b}$). Οι αριθμοί - by default - αυξάνονται κατά 1 αλλά αυτό είναι κάτι που μπορούμε να το αλλάξουμε, αν θέλουμε. Ο παραπάνω επαναληπτικός βρόχος δίνει στη μεταβλητή $\textrm{i}$ τις τιμές της ακολουθίας $1,2,3,4, \cdots, n$, μια προς μια, σε κάθε επανάληψη.

Ας τη δοκιμάσουμε:

```python
print("Example 1 : 0! = {}".format(factorial(0)))     # Πρέπει να επιστρέφει 0! = 1
print("Example 2 : 1! = {}".format(factorial(1)))     # Πρέπει να επιστρέφει 1! = 1
print("Example 3 : 5! = {}".format(factorial(5)))     # Πρέπει να επιστρέφει 5! = 120
```

Βλέπουμε ότι τα αποτελέσματα της είναι σωστά.

Αν αφήσουμε μια μπάλα από ύψος $h$, αυτή επιταχύνεται προς τα κάτω λόγω της επιτάχυνσης της βαρύτητας με μέτρο $9.8$ m/s $^2$. Αν την πετάξουμε κατακόρυφα προς τα κάτω με κάποια αρχική ταχύτητα $\vec{u}_{0y} = \vec{u}_{0} = -u_0
\vec{j}$, περιμένουμε να φτάσει πιο γρήγορα στο έδαφος. 

Θα γράψουμε ένα πρόγραμμα που δέχεται το ύψος $h$ (θέση Β στο σχήμα) και την κατακόρυφη προς τα κάτω αρχική ταχύτητα $u_0$ (σε μέτρο) και τυπώνει τη γραφική παράσταση της θέσης, της ταχύτητας, και της επιτάχυνσης της μπάλας μέχρι αυτή να χτυπήσει στο έδαφος (θέση Α στο σχήμα). Αρχή του συστήματος αναφοράς μας θα είναι το έδαφος (δηλ. η συμβολή των αξόνων βρίσκεται στο σημείο Α) και θετική φορά της κίνησης θα είναι προς τα επάνω.

Θα μελετήσουμε τη διαδρομή B $\to$ A, δηλ. αρχική θέση το σημείο Β και τελική θέση το σημείο Α. Ο χρόνος που απαιτείται για να φτάσει η μπάλα στο έδαφος δίνεται από τη γενική σχέση 

$$y_f = y_i + u_{0y}t + \frac{1}{2}a_y t^2$$ 

Λόγω των παραδοχών που κάναμε (θετική φορά προς τα πάνω, και διανύσματα $\vec{u}_{0_y}$ και $\vec{g}$ με φορά προς τα κάτω) και της διάταξης του προβλήματος, θα έχουμε

$$y_A = y_B - u_{0_{B}}t - \frac{1}{2}gt^2 \Longleftrightarrow 0 = h - u_{0_{B}}t - \frac{1}{2}gt^2 \Longleftrightarrow \frac{1}{2}gt^2 + u_{0_{B}}t - h = 0$$

Αν και μπορούμε να βρούμε το θετικό $t$ από τη γνωστή για τα δευτεροβάθμια πολυώνυμα σχέση 

$$ \frac{-\beta \pm \sqrt{\beta^2 - 4\alpha \gamma}}{2\alpha}$$

ας δούμε παρακάτω πώς μπορεί η Python να μας βρει τη θετική ρίζα και πώς μπορούμε μετά να υπολογίσουμε θέση, ταχύτητα, και επιτάχυνση για το πρόβλημά μας, για κάθε χρονική στιγμή $t$ της διαδρομής ΒΑ.

***

## Ζητούμενο:

Συμπληρώστε τον παρακάτω κώδικα.

```python

def free_fall_functions(h, u0y):
    
    # Συνάρτηση που υπολογίζει θέση, ταχύτητα, και επιτάχυνση στη διαδρομή Β -> Α παραπάνω.
    # Mεταβλητή h: ύψος ρίψης.
    # Mεταβλητή u0y: μέτρο αρχικής ταχύτητας ρίψης.
    # Επιστρέφει:
    # - Θέση y(t)
    # - Ταχύτητα u(t)
    # - Επιτάχυνση a(t)
    # - Xρόνος t
    
    # Συνθήκες: 
    g = 9.8     # Μέτρο επιτάχυνσης της βαρύτητας.
    
    yB = h     # Αρχική θέση.
    
    # Χρόνος μέχρι η μπαλα να χτυπήσει το έδαφος - ζητάμε να βρούμε τις λύσεις του 2οβάθμιου πολυωνύμου που βρηκαμε παραπάνω:
    tA = np.roots([0.5*g, u0y, -h])   # η ".roots" δέχεται συντελεστές πολυωνύμου και επιστρέφει τις ρίζες του πολυωνύμου!
    
    # Κρατάμε μόνο τη θετική λύση (δεν υπάρχει αρνητικός χρόνος):
    tA = tA[tA > 0]
    
    # Κατασκευάζουμε τον άξονα του χρόνου όπου θα πάρουμε τις μετρήσεις μας (από θέση Β, όπου t = 0 ως θέση Α, όπου t = tA που βρήκαμε παραπάνω):
    tB = 0
    t = np.linspace(tB, tA, 1000)
    
    # Θέση, ταχύτητα, επιτάχυνση (επιστρέφονται στο χρήστη, μαζί με το διάνυσμα t) - προσέξτε τα πρόσημα στις υλοποιήσεις!
    y = yB + u0y * t - 1 / 2 * g * t ** 2     # Συμπληρώθηκε (χρησιμοποιώντας τα yB, u0y, g και t)!
    u = u0y - g * t                           # Συμπληρώθηκε (χρησιμοποιώντας τα u0y, g και t)!
    a = - g                                   # Συμπληρώθηκε (χρησιμοποιώντας το g)!
    
    return y, u, a, t
    
```

Παρατηρήστε ότι καλέσαμε τη συνάρτηση $\textrm{roots}$ της βιβλιοθήκης numpy για να μας επιστρέψει τις λύσεις της εξίσωσης 

$$\frac{1}{2}gt^2 + u_{0_{B}}t - h=0$$

Από τις δυο ρίζες που μας επιστρέφει η συνάρτηση, κρατάμε τη θετική, μια και μιλάμε για χρόνο.

Ας τυπώσουμε τις γραφικές παραστάσεις για $h=10$ m χωρίς αρχική ταχύτητα και για $h=50$ m με αρχική ταχύτητα μέτρου $u_0 = 5$ m/s:

```python

y, u, a, t = free_fall_functions(10, 0)     # Παράδειγμα: h = 10, u0 = 0.

plt.figure(figsize=(8, 10)) 
plt.subplot(3, 1, 1)
plt.plot(t, y)
plt.title('Θέση')
plt.subplot(3, 1, 2)
plt.plot(t, u)
plt.title('Ταχύτητα')
plt.subplot(3, 1, 3)
plt.plot(t, a * np.ones(np.size(t)))
plt.title('Επιτάχυνση')
plt.show()                                  # Εμφανίζουμε την γραφική παράσταση και τελειώσαμε!
```

```python

y, u, a, t = free_fall_functions(50, 5)     # Παράδειγμα: h = 50, u0 = 5.

plt.figure(figsize=(8, 10)) 
plt.subplot(3, 1, 1)
plt.plot(t, y)
plt.title('Θέση')
plt.subplot(3, 1, 2)
plt.plot(t, u)
plt.title('Ταχύτητα')
plt.subplot(3, 1, 3)
plt.plot(t, a * np.ones(np.size(t)))
plt.title('Επιτάχυνση')
plt.show()                                  # Εμφανίζουμε την γραφική παράσταση και τελειώσαμε!
```

Αν τα κάνατε όλα σωστά θα πρέπει τα γραφήματα που προκύπτουν παραπάνω να είναι τα αναμενόμενα: στην επιταχυνόμενη με **σταθερή** επιτάχυνση κίνηση: 

- Η θέση είναι **δευτεροβάθμια συνάρτηση ως προς t** και μειώνεται με την πάροδο του χρόνου (λαμβάνοντας υπ'όψη μας το αρνητικό της πρόσημο λόγω φοράς διανύσματος θέσης)
- η ταχύτητα είναι **γραμμική** και μειώνεται με την πάροδο του χρόνου (λαμβάνοντας υπ'όψη μας το αρνητικό της πρόσημο λόγω φοράς διανύσματος ταχύτητας), και 
- η επιτάχυνση είναι **σταθερή** και ίση με $-9.8$ m/s $^2$!

# Άσκηση Python 2

Έστω τώρα ότι εκτοξεύουμε την μπάλα από ύψος $h$ με αρχική ταχύτητα $u_0$ υπό γωνία $\theta$! Θέλουμε να υπολογίσουμε την οριζόντια απόσταση που θα διανύσει η μπάλα μέχρι να χτυπήσει στο έδαφος. 

Προσέξτε! Αυτή η απόσταση ΔΕΝ είναι το εύρος βολής $R$ (ή αλλιώς, το βεληνεκές της)! Το εύρος βολής μας δίνει την οριζόντια απόσταση από το σημείο ρίψης ως το σημείο που η μπάλα επιστρέφει στο ίδιο επίπεδο με το σημείο ρίψης (θέση Β)! Στο παραπάνω πρόβλημα **ΔΕ** ζητείται αυτό!



Στην παραπάνω εικόνα, $R$ είναι το εύρος της βολής. Η απόσταση που ζητείται στο δικό μας πρόβλημα συμβολίζεται με $d$ !

Ας λύσουμε το πρόβλημα θεωρητικά ώστε να ξέρουμε τις εξισώσεις μας. Θεωρούμε τις συμβατικές θετικές φορές (πάνω, δεξιά). Σημείο αναφοράς μας είναι το έδαφος $h$ μέτρα κάτω από το σημείο ρίψης της μπάλας, οπότε το σημείο $A$ έχει συντεταγμένες $(x_A, y_A) = (0, h)$ και το σημείο $C$ έχει συντεταγμένες $(x_C , y_C) = (d, 0)$ (και σε περίπτωση που το σκεφτήκατε, το σημείο $B$ έχει συντεταγμένες $(x_B, y_B) = (R, h)$).

Η κίνηση της μπάλας μπορεί να περιγραφεί από το μοντέλο της ευθύγραμμης κίνησης με σταθερή ταχύτητα στον άξονα $x'x$ (δεν επιταχύνεται στον άξονα αυτό) και από το μοντέλο της ευθύγραμμης επιταχυνόμενης κίνησης με σταθερή επιτάχυνση στον άξονα $y'y$, με επιτάχυνση $\vec{a}_y = -\vec{g} = (-9.8 \text{ m/s}^2)\vec{j}$.

Από το σημείο $A$ ως το σημείο $C$, στην οριζόντια συνιστώσα της κίνησης, ισχύει 

$$x_C = x_A + u_{0x}t \Longleftrightarrow d = 0 + u_0 \cos(\theta) t = u_0\cos(\theta)t$$

Στην ίδια διαδρομή, στην κατακόρυφη συνιστώσα της κίνησης ισχύει

$$y_C = y_A + u_{0y}t - \frac{1}{2}gt^2 \Longleftrightarrow 0 = h + u_0\sin(\theta)t - \frac{1}{2}gt^2 \Longleftrightarrow gt^2 - 2u_0\sin(\theta)t - 2h = 0$$

Έχουμε πάλι δευτεροβάθμιο πολυώνυμο και θα μπορούσαμε πάλι να "φωνάξουμε" τη συνάρτηση $\textrm{roots}$ της numpy να μας βρει τη θετική λύση. Αλλά ας σας παιδέψουμε λίγο. 😃 Λύνοντας αναλυτικά,

$$t = \frac{2u_0\sin(\theta) + \sqrt{4u_0^2\sin^2(\theta) + 8gh}}{2g} $$

και άρα

$$d = u_0\cos(\theta)t = u_0\cos(\theta)\left( \frac{2u_0\sin(\theta) + \sqrt{4u_0^2\sin^2(\theta) + 8gh}}{2g}\right) = \left( \frac{u_0^2\sin(2\theta) + \sqrt{u_0^4\sin^2(2\theta) + 8ghu_0^2\cos^2(\theta)}}{2g}\right)$$

***

## Ζητούμενο 1:


Γράψτε συνάρτηση

η οποία υπολογίζει και επιστρέφει τη ζητούμενη απόσταση $d$.

```python

def twoD_motion_d(h, u0, theta):
    
    # Συνάρτηση που υπολογίζει την απόσταση d όπως ορίστηκε παραπάνω.
    # Μεταβλητή h: ύψος ρίψης.
    # Μεταβλητή u0: μέτρο αρχικής ταχύτητας ρίψης.
    # Μεταβλητή theta: αρχική γωνία ρίψης.
    # Επιστρέφει:
    # - Απόσταση d

    # Γράψτε εδώ τον κώδικά σας - προτείνω να μη γράψετε την ζητούμενη εξίσωση σε μια γραμμή, "σπάστε τη" σε όρους για ευκολία...
    
    g = 9.81     # Σταθερά g - επιτάχυνση της βαρύτητας.
    term1 = u0 ** 2 * np.sin(2 * theta)     # Πράξεις αριθμητή και αριστερά του +.
    term2 = np.sqrt(u0 ** 4 * np.sin(2 * theta) ** 2 + 8 * g * h * u0 ** 2 * np.cos(theta) ** 2)     # Πράξεις αριθμητή και δεξιά του +.
    term3 = 2 * g     # Πράξεις παρανομαστή.
    d = ((term1 + term2) / term3)     # Συνδυασμοί - ολόκληρη η παράσταση d.
    return d
    
    # Τέλος κώδικα!
    
```

Εκτελέστε την παρακάτω γραμμή για επιβεβαίωση:

```python
print(twoD_motion_d(10, 5, np.pi/4)) # Παράδειγμα επιβεβαίωσης: h=10, u0=5, θ=π/4, πρέπει να πάρετε τιμή κοντά στο 6.4848!
```

## Ζητούμενο 2:

Καταλήξαμε σε μια αλγεβρική σχέση που να συνδέει τις μεταβλητές $u_0, \theta, h, d$ σε μια βολή από ύψος $h$. 

Αν όμως ζητούνταν να βρούμε τη **γωνία** ρίψης $\theta$ ώστε η μπάλα να πετύχει ένα στόχο σε οριζόντια απόσταση $D$, με δοσμένη αρχική ταχύτητα $u_0$ και δοσμένο ύψος ρίψης $h$, τι θα κάνατε? Είναι εφικτό να λύσουμε την παραπάνω εξίσωση ως προς $\theta$ αλλά είναι **ιδιαίτερα επίπονο** (και ήδη σας ταλαιπωρήσαμε με το προηγούμενο ζητούμενο 😃)... αν δεν το πιστεύετε, δείτε: πόσο εύκολο είναι να λύσετε ως προς $\theta$ την προηγούμενη εξίσωση?

$$d = \left( \frac{u_0^2\sin(2\theta) + \sqrt{u_0^4\sin^2(2\theta) + 8ghu_0^2\cos^2(\theta)}}{2g}\right)$$

Δε μοιάζει καθόλου μα καθόλου εύκολο... μπορούμε όμως να προσεγγίσουμε το παραπάνω πρόβλημα **αριθμητικά**. Ας ορίσουμε μια συνάρτηση 

$$F(\theta) = d(\theta) - D$$

με $d(\theta)$ τη συνάρτηση που βρήκαμε στο προηγούμενο ζητούμενο, και $D$ τη δοσμένη οριζόντια απόσταση. 

Αν τυπώσουμε τη γραφική παράσταση της $F(\theta)$ ως προς $\theta$ τότε μπορούμε να προσεγγίσουμε γραφικά την τιμή της $\theta$ για την οποία 

$$F(\theta) = 0 \Longleftrightarrow d(\theta) = D$$

Με άλλα λόγια, η τιμή της $\theta$ για την οποία $F(\theta) = 0$ είναι η ζητούμενη γωνία. Όμως αυτή τη φορά δε θα τη βρούμε αναλυτικά αλλά από τη γραφική παράσταση. 

Συμπληρώστε την παρακάτω συνάρτηση η οποία αναλαμβάνει τα παραπάνω, δηλ. υλοποιεί τη συνάρτηση $F(\theta)$.

```python
def find_theta(D, h, u0):
    # Συνάρτηση που υπολογίζει τη γωνία θ για την οποία F(θ) = 0.
    # Μεταβλητή h: ύψος ρίψης.
    # Μεταβλητή D: οριζόντια απόσταση από το σημείο ρίψης.
    # Μεταβλητή u0: μέτρο αρχικής ταχύτητας.
    # Επιστρέφει:
    # - Τιμές συνάρτησης F για κάθε θ από 0 ως π/2.
    # - Διάνυσμα τιμών θ στο διάστημα [0, π/2).

    g = 9.8    # Βαρυτική επιτάχυνση - κατά μέτρο.

    theta = np.linspace(0, np.pi / 2, 1000)    # Πιθανές τιμές της θ στο [0, π/2) εκτιμώμενες σε 1000 σημεία.

    # Από το προηγούμενο ερώτημα έχουμε το d:
    term1 = u0 ** 2 * np.sin(2 * theta)
    term2 = np.sqrt(u0 ** 4 * np.sin(2 * theta) ** 2 + 8 * g * h * u0 ** 2 * np.cos(theta) ** 2)
    term3 = 2 * g
    d = (term1 + term2) / term3

    # Και έτσι βρίσκουμε και το F:
    F = d - D
    
    return F, theta
    
# Τέλος κώδικα
```

Ας βάλουμε τις παραμέτρους του προηγούμενου ζητούμενου για να δούμε αν θα βρούμε την τιμή $\pi/4$, που ήταν η γωνία του προηγούμενου ζητουμένου!

```python
# Στο προηγούμενο ζητουμενο βάλαμε h=10, u0=5, και βρήκαμε d=6.4848. Τώρα θέλουμε να βρούμε το θ (που πριν το είχαμε βάλει ίσο με π/4)!
F, theta = find_theta(6.4848, 10, 5)            

plt.figure(figsize=(14, 8)) 
plt.plot(theta, F, np.pi / 4, 0, 'rX')
plt.plot(theta, 0 * np.ones(np.size(theta)), 'k--')
plt.plot((np.pi / 4, np.pi / 4), (-6, 2),  'k--', scaley = False)
plt.show()                                  # Εμφανίζουμε την γραφική παράσταση και τελειώσαμε!
```

Αν τα κάνατε όλα σωστά, πρέπει να βλέπετε ότι πράγματι η γραφική παράσταση μηδενίζεται στη γωνία $\pi/4 = 0.7854$ για τις παραμέτρους που θέσαμε!!

# Άσκηση Python 3

Στο μάθημα είδαμε ότι συμπληρωματικές γωνίες (γωνίες με άθροισμα $90$ μοίρες) ρίψης δίνουν το ίδιο εύρος (ή βεληνεκές) σε μια βολή, όταν αυτή εκτελείται με το ίδιο μέτρο αρχικής ταχύτητας. Ας δούμε κατ'αρχάς το γιατί.

Ξέρουμε ότι

$$R = \frac{u_i^2 \sin(2\theta_i)}{g}$$

με $u_i$ η αρχική ταχύτητα ρίψης και $\theta_i$ η γωνία ρίψης. Μπορούμε να το αποδείξουμε αλγεβρικά, δηλ.

$$R' = \frac{u_i^2 \sin\left(2\left(\frac{\pi}{2} - \theta_i\right)\right)}{g} = \frac{u_i^2 \sin(\pi - 2\theta_i)}{g} = \frac{u_i^2 \sin(2\theta_i)}{g} = R$$ 

αφού $\sin(\pi - x) = \sin(x)$, μπορούμε όμως να το αποδείξουμε και γεωμετρικά. Δείτε το παρακάτω σχήμα.




Στην άσκηση αυτή θα γράψουμε μια συνάρτηση που να μας κάνει το γράφημα της **τροχιάς** της βολής στο $xy$ επίπεδο και μέσω αυτού θα επιβεβαιώσουμε την ιδιότητα των συμπληρωματικών γωνιών.

Όμως όλες οι εξισώσεις που γνωρίζουμε είναι συναρτήσεις είτε $x(t)$ είτε $y(t)$! Δεν έχουμε κάποια συνάρτηση $y(x)$ για να θέσουμε $x=0, 0.01, 0.02, \cdots, R$ και να εμφανίσουμε το γράφημα $y(x)$! 

Ε, ας τη βρούμε λοιπόν! Για ευκολία, υποθέτουμε ότι σημείο βολής είναι η συμβολή των αξόνων, $(0,0)$ και θεωρούμε θετικές φορές της κίνησης τις συμβατικές φορές των αξόνων (πάνω + δεξιά).

Για τη διαδρομή από το σημείο ρίψης $(x_i, y_i) = (0,0)$ μέχρι ένα τυχαίο σημείο της τροχιάς $(x,y)$, θα δουλέψουμε κατά άξονες. Γνωρίζουμε ότι στον άξονα $x'x$ της κίνησης, η ταχύτητα είναι σταθερή:

$$x = x_i + u_{x}t \Longrightarrow x = 0 + u_i \cos(\theta_i) t \Longrightarrow t = \frac{x}{u_i \cos(\theta_i)}$$

Στον άξονα $y'y$ της κίνησης, η κίνηση είναι επιταχυνόμενη με σταθερή επιτάχυνση (αυτή της βαρύτητας), οπότε

$$y = y_i + u_{y_i}t - \frac{1}{2}gt^2 \Longrightarrow y = 0 + u_i \sin(\theta_i)t - \frac{1}{2}gt^2$$

και αντικαθιστώντας

$$y = u_i\sin(\theta_i)\frac{x}{u_i\cos(\theta_i)} - \frac{1}{2}g\left(\frac{x}{u_i\cos(\theta_i)}\right)^2$$

Απλοποιώντας,

$$y(x) = x\tan(\theta_i) - \frac{gx^2}{2u_i^2 \cos^2(\theta_i)}$$

## Ζητούμενο:

Γράψτε συνάρτηση

η οποία υλοποιεί και επιστρέφει την τροχιά της βολής μέχρι το σημείο επαναφοράς στο επίπεδο που ορίζεται από τον άξονα $x'x$ (δηλ. μέχρι $x=R$).

```python
def projectile_track(ui, theta_i):
    # Συνάρτηση που υπολογίζει την τροχιά μιας βολής, δηλ. μια συνάρτηση y(x).
    # Mεταβλητή theta_i: αρχική γωνία ρίψης.
    # Mεταβλητή ui: μέτρο αρχικής ταχύτητας ρίψης.
    # Επιστρέφει:
    # - Τροχιά y(x) για x στο [0, R)
    # - Τιμές του άξονα x'x στο [0, R)

    # Σταθερές:
    g = 9.8
    
    # Εύρος (όπως το ξέρετε από τη θεωρία) - συμπληρωμένο:
    R = (ui ** 2 * np.sin(2 * theta_i)) / g
    
    # Άξονας x'x από 0 ως R εκτιμώμενος σε 1000 σημεία:
    xx = np.linspace(0, R, 1000)
    
    # Τροχιά y ως προς xx, όπως το δείξαμε παραπάνω - συμπληρωμένο:
    term1 = (xx * np.tan(theta_i))                   # 1. x⋅tan(θ1)
    term2 = (g * xx ** 2)                            # 2. g⋅x^2
    term3 = (2 * ui ** 2 * np.cos(theta_i) ** 2)     # 3. 2⋅ui^2⋅cos^2(θ1)
    y = term1 - term2 / term3
    
    return y, xx
# Τέλος κώδικα
```

Ας την ελέγξουμε !!

```python
theta = np.pi / 3
y, x = projectile_track(40, theta)                 # Γωνία 60 μοίρες, αρχική ταχύτητα 40m/s.
plt.figure(figsize=(10, 4))
plt.plot(x, y)
y, x = projectile_track(40, np.pi / 2 - theta)     # Γωνία 90-60=30 μοίρες, συμπληρωματική, αρχική ταχύτητα 40m/s.
plt.plot(x, y)                                     # Πρέπει το εύρος βολής να είναι το ίδιο! (λίγο πάνω από 140 μ.).
plt.grid()
plt.show()                                         # Εμφανίζουμε την γραφική παράσταση και τελειώσαμε!
```
