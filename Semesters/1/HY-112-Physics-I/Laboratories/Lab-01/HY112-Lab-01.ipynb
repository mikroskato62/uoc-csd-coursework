# Άσκηση Python 1

Μιας και στη θεωρία έχουμε δει διανύσματα, ας μάθουμε έναν απλό τρόπο να "παίζουμε" στην Python με αυτά. Ας θεωρήσουμε τα διανύσματα 

\begin{align}
\vec{a} &= 2\vec{i}-2\vec{j} \nonumber \\
\vec{b} &= -2\vec{i} + 4\vec{j} \nonumber 
\end{align}

Πως μπορούμε να τα αναπαραστήσουμε στην Python? Δείτε (επιλέξτε το παρακάτω κομμάτι κώδικα και πατήστε Shift+Enter):

```python
a = np.array([2,-2])                  # ορίζουμε ένα διάνυσμα με συντεταγμένες (2,-2)
b = np.array([-2, 4])                 # ορίζουμε ένα διάνυσμα με συντεταγμένες (-2,4)
```

Οι παραπάνω μεταβλητές a, b αποτελούν πίνακες (arrays) και περιέχουν δυο στοιχεία η καθεμία. Θα υποθέσουμε ότι το πρώτο στοιχείο αντιστοιχεί στη $x$-συνιστώσα του εκάστοτε διανύσματος, ενώ το δεύτερο στοιχείο στην $y$-συνιστώσα του. Καθένα στοιχείο προσπελάζεται μέσω δείκτη ως: a[0], a[1], κλπ.

Μπορούμε με αυτά να ορίσουμε σύνθετα διανύσματα με τις γνωστές πράξεις μεταξύ διανυσμάτων. Ας ορίσουμε τα 

\begin{align}
\vec{a}&+\vec{b} \nonumber \\
\vec{a}&-\vec{b}\nonumber 
\end{align}

```python
aplusb = a + b
aminusb = a - b
```

Παρ'όλο που μια "σκέτη" εντολή print θα σας δείξει τα περιεχόμενα των παραπάνω διανυσμάτων - με το πρώτο στοιχείο να είναι η $x$-συνιστώσα και το δεύτερο στοιχείο να είναι η $y$-συνιστώσα - ας δούμε έναν πιο όμορφο τρόπο να παρουσιάζουμε τα διανύσματα.

```python
print("Διάνυσμα a + b = {}i + {}j".format(aplusb[0], aplusb[1]))
print("Διάνυσμα a - b = {}i + {}j".format(aminusb[0], aminusb[1]))
```

Μπορούμε τώρα να υπολογίσουμε το μέτρο αυτών των διανυσμάτων. Θυμηθείτε τον ορισμό: Αν $\vec{a} = a_x\vec{i} +  a_y\vec{j}$, τότε 

\begin{equation} 
|\vec{a}| = \sqrt{a_x^2 + a_y^2} \nonumber 
\end{equation}

```python
abs_a = np.sqrt(np.sum(a**2))           # καλούμε τη συνάρτηση np.sum για να προσθέσει τα τετράγωνα των συντεταγμένων
                                        # και τη συνάρτηση np.sqrt για να βρει την τετρ. ρίζα του αθροίσματος
abs_b = np.sqrt(np.sum(b**2))                # όμοια ακριβώς
abs_aplusb = np.sqrt(np.sum(aplusb**2))      # όμοια ακριβώς
abs_aminusb = np.sqrt(np.sum(aminusb**2))    # όμοια ακριβώς

print("Μέτρα: |a| = {}, |b| = {}, |a+b| = {}, |a-b| = {}".format(abs_a, abs_b, abs_aplusb, abs_aminusb))
```

Ας κάνουμε το αντίστοιχο για τις γωνίες. Θυμηθείτε ότι αν $\vec{a} = a_x\vec{i} + a_y\vec{j}$ τότε 

\begin{equation}
\angle \vec{a} = \tan^{-1}\left(\frac{a_y}{a_x}\right) \nonumber 
\end{equation}

Ευτυχώς η Python έχει έτοιμη συνάρτηση, την $\textrm{arctan2(y,x)}$, η οποία σας επιστρέφει σωστά τη γωνία του διανύσματος, χωρίς να χρειαστεί να προσθέσετε/αφαιρέσετε $\pm \pi$ στο αποτέλεσμά σας, όπως είδατε στη θεωρία. Τέλος, θα εκφράσουμε τα αποτελέσματά μας σε μοίρες αντί για radians. Εκτελέστε το παρακάτω κελί, επιλέγοντάς το και πατώντας Shift + Enter.

```python
theta1 = np.arctan2(a[1], a[0])
theta2 = np.arctan2(b[1], b[0])
theta3 = np.arctan2(aplusb[1], aplusb[0])
theta4 = np.arctan2(aminusb[1], aminusb[0])
```

Ας τυπώσουμε αυτές τις γωνίες σε μοίρες. Θυμηθείτε ότι 
\begin{equation}
\frac{deg}{180} = \frac{rad}{\pi} \nonumber 
\end{equation}

```python
print("Γωνίες: <a = {}, <b = {}, <(a+b) = {}, <(a-b) = {}".
      format(theta1*180/np.pi, theta2*180/np.pi, theta3*180/np.pi, theta4*180/np.pi))
```

\begin{equation}
\vec{c} = 3\vec{a} - 6\vec{b} \nonumber

Υπολογίστε το μέτρο και τη γωνία του διανύσματος αυτού.

# Άσκηση Python 2

Ωραία, τώρα έχετε καταλάβει βασικούς χειρισμούς των διανυσμάτων στην Python. Όμως μάθαμε στη θεωρία και μια άλλη πολύ σημαντική πράξη μεταξύ διανυσμάτων: το $\textbf{εσωτερικό γινόμενο}$!

Πώς μπορούμε να το υπολογίσουμε αυτό?

Ας χρησιμοποιήσουμε πάλι τα γνωστά μας διανύσματα $$\vec{a} = 2\vec{i}-2\vec{j}$$ και $$\vec{b} = -2\vec{i} + 4\vec{j}$$

```python
a = np.array([2,-2])
b = np.array([-2, 4])
```

Η θεωρία μας πληροφορεί ότι το εσωτερικό γινόμενο δίνεται ως 

$$\vec{a}\cdot \vec{b} = |\vec{a}||\vec{b}|\cos(\theta) = a_xb_x + a_yb_y$$

Μια και έχουμε τις συνιστώσες, ας τις αξιοποιήσουμε. Έτσι, μια "αφελής" προσέγγιση θα ήταν η εξής:

```python
c = a[0]*b[0] + a[1]*b[1]      # συμφωνήσαμε ότι το πρώτο στοιχείο του array θα είναι η x-συνιστώσα και το δεύτερο η y-συνιστώσα

print("Εσωτερικό γινόμενο a.b = {}".format(c))
```

Αν όμως το διάνυσμά μας είχε $200$ συνιστώσες και όχι $2$, θα έπρεπε είτε να γράψουμε μια ΠΟΛΥ μεγάλη γραμμή κώδικα, είτε να χρησιμοποιήσουμε ένα βρόχο επανάληψης (for loop, π.χ.) για να "μαζέψουμε" όλα τα γινόμενα.

Ευτυχώς, η numpy έχει έτοιμη συνάρτηση που μπορεί να υπολογίζει εσωτερικά γινόμενα για διανύσματα οσοδήποτε μεγάλων διαστάσεων! Η συνάρτηση αυτή ονομάζεται $\textrm{dot( )}$. Ας τη δούμε:

```python
c = np.dot(a,b)

print("Εσωτερικό γινόμενο a.b = {}".format(c))
```

## Ζητούμενο:

Τρία διανύσματα δίνονται ως 
\begin{align}
\vec{a} &= \vec{i} + 4\vec{j} + 7\vec{k} \nonumber \\
\vec{b} &= \vec{j} - \vec{k} \nonumber \\
\vec{c} &= -\vec{i} + 2\vec{j} + \vec{k} \nonumber 
\end{align}

με $\vec{i}, \vec{j}, \vec{k}$ να είναι τα μοναδιαία διανύσματα του τριδιάστατου χώρου. Υπολογίστε και τυπώστε τα:

\begin{align}
(α)& \: \vec{a}\cdot (\vec{b}+\vec{c}) \nonumber \\
(β)& \: \vec{a} \cdot \vec{b} - \vec{a}\cdot \vec{c} \nonumber \\
(γ)& \: (\vec{a}\cdot \vec{c})(\vec{b}\cdot \vec{c})\nonumber 
\end{align}


```python
# Γράψτε εδώ τον κώδικά σας

# ------------------------------------------------ 2 ------------------------------------------------ #

import numpy as np                                       # Εισαγωγή βιβλιοθήκης numpy ως np.

a = np.array([1, 4, 7])                                  # Προκαθορισμένο διάνυσμα α.
b = np.array([0, 1, -1])                                 # Προκαθορισμένο διάνυσμα b.
c = np.array([-1, 2, 1])                                 # Προκαθορισμένο διάνυσμα c.

abc1 = np.dot(a, (b + c))                                # Εύρεση εσωτερικού γινομένου παράστασης (α).
abc2 = np.dot(a, (b - c))                                # Εύρεση εσωτερικού γινομένου παράστασης (β).
abc3 = np.dot(np.dot(a, c), np.dot(b, c))                # Εύρεση εσωτερικού γινομένου παράστασης (γ).

print(f"\n⋅ Εσωτερικά γινόμενα: "
      f"\n(α): a ⋅ (b + c) = {abc1} "
      f"\n(β): a ⋅ b - a ⋅ c = {abc2} "
      f"\n(γ): (a ⋅ c) ⋅ (b ⋅ c) = {abc3}\n")            # Εκτύπωση δεδομένων.

# ------------------------------------------------ 2 ------------------------------------------------ #

# Τέλος κώδικα
```

# Άσκηση Python 3

Σε αυτήν την άσκηση θα θυμηθούμε λίγο τα Μαθηματικά του Λυκείου (ή του Απειροστικού Λογισμού) και μάθουμε πως να υπολογίζουμε αριθμητικά ορισμένα ολοκληρώματα! Γνωρίζετε ότι ένα ορισμένο ολοκλήρωμα υπολογίζεται ως 

$$\int_a^b f(x)dx = F(b) - F(a)$$ 

με $F'(x) = f(x)$. Εμείς θα χρησιμοποιήσουμε μια λίγο πιο... πρακτική προσέγγιση. 😃

Προφανώς η πράξη της ολοκλήρωσης γίνεται σε ένα συνεχές διάστημα $[a,b]$. Στον υπολογιστή, δεν μπορούμε να αναπαραστήσουμε αριθμητικά κάθε σημείο του διαστήματος αυτού (χρειαζόμαστε άπειρα σημεία και άρα άπειρη μνήμη). Αναγκαστικά θα πρέπει να $\textit{δειγματοληπτήσουμε}$ το διάστημα αυτό, όσο πυκνά ή αραιά θέλουμε. Δειγματοληψία δε σημαίνει τίποτε άλλο από το να πάρουμε τιμές του διαστήματος $[a,b]$. Φυσικά, όσο πυκνότερη η δειγματοληψία τόσο πιο "πιστή" η αναπαράσταση του συνεχούς διαστήματος $[a,b]$.

Θα μάθετε στον Απειροστικό Λογισμό σας (οπότε **μη δώσετε ιδιαίτερη σημασία στις τεχνικές λεπτομέρειες εδώ**, οι οποίες θα είναι οι ελάχιστες δυνατές) ότι ένα ολοκλήρωμα μπορεί να προσεγγιστεί αριθμητικά ως εξής: 

$$ \int_a^b f(x)dx \approx \sum_{i=0}^n f(ξ_i)\Delta x_i$$

με $\Delta x_i$ να είναι το $i$-οστό υποδιάστημα του $[a,b]$ και $ξ_i$ το κέντρο κάθε υποδιαστήματος. Αυτό σημαίνει ότι πρέπει να διαμερίσουμε το διάστημα $[a,b]$ σε υποδιαστήματα: ομοιόμορφα ή μη. Ας πάρουμε την απλή προσέγγιση της ομοιόμορφης διαμέρισης, όπου $\Delta x_i = x_{i}-x_{i-1} = \Delta x$. Τότε 

$$a = {x_0} \lt {x_1} \lt {x_2} \lt \cdots \lt {x_i} \lt \cdots \lt {x_n} = b$$

και έτσι τα υποδιαστήματα θα είναι

$$\left[ {{x_0},{x_1}} \right],\left[ {{x_1},{x_2}} \right], \ldots , \left[ {{x_{n - 1}},{x_n}} \right]$$

με $x_0 = a, x_n = b$ και κάθε υποδιάστημα είναι ίδιου "πλάτους" με οποιοδήποτε άλλο. Το τελευταίο βήμα είναι να θεωρήσουμε ότι η τιμή της συνάρτησης $f(x)$ σε κάθε υποδιάστημα είναι $\textit{σταθερή}$ και ίση με $f(ξ_i)$ σε $\textbf{όλο}$ το υποδιάστημα $\Delta x_i$, με $ξ_i = \frac{x_i - x_{i-1}}{2}$.

Επειδή γνωρίζετε ότι το ολοκλήρωμα δεν είναι τίποτε άλλο από την "επιφάνεια κάτω από τη συνάρτηση" $f(x)$, αυτό που καταφέραμε να κάνουμε με την παραπάνω διαδικασία είναι να προσεγγίσουμε την "επιφάνεια" αυτή από "παραλληλόγραμμα" (ας τα ονομάσουμε $R_i$), όπως στο σχήμα:



Βλέπετε στο σχήμα τα κέντρα υποδιαστημάτων $ξ_i$, τη διαμέριση με τα υποδιαστήματα $\Delta x_i$, και τη σταθερή τιμή και ίση με $f(ξ_i)$ της συνάρτησης $f(x)$ που θεωρούμε ότι παίρνει σε κάθε υποδιάστημα $\Delta x_i$. Παρατηρήστε επίσης ότι οι αποστάσεις μεταξύ των $ξ_i$ είναι σταθερές!

Όπως καταλαβαίνετε, τα παραλληλόγραμμα $R_i$ έχουν συνολική επιφάνεια που προσεγγίζει την "επιφάνεια κάτω από τη συνάρτηση" $f(x)$. Όσο πιο μικρού πλάτους $\Delta x_i$ τα παραλληλόγραμμα, τόσο καλύτερη η προσέγγιση! Ιδανικά, αν $\Delta x_i \to 0$, τότε το άθροισμα των επιφανειών των παραλληλογράμμων ισούται με το ολοκλήρωμα της συνάρτησης $f(x)$!

Ας υποθέσουμε ότι έχουμε το ολοκλήρωμα $$\int_0^1 \frac{2x}{\sqrt[3]{x^2 + 1}}dx$$

Αναγνωρίζουμε ότι $[a,b] = [0,1]$ και ότι $f(x) = \frac{2x}{\sqrt[3]{x^2 + 1}}$. Κοιτάξτε πόσο εύκολα μπορούμε να το υπολογίσουμε:

```python
x = np.linspace(0,1,100)                  # διαμερίζουμε ομοιόμορφα το διάστημα [0,1] σε 100 σημεία και τα αποθηκεύουμε στο x!
                                          # Αυτά είναι τα ξ_i μας!
dx = x[1]-x[0]                            # λόγω ομοιόμορφης διαμέρισης, όλα τα διαστήματα Δx_i είναι ίσα με Δx_1 = x[1]-x[0]

func1 = 2*x/(x**2 + 1)**(1/3)              # ορίζουμε τη συνάρτηση f(x) και την υπολογίζουμε πάνω στην ομοιόμορφη διαμέριση x
Result1 = dx*np.sum(func1)                 # εκτελούμε το άθροισμα-προσέγγιση του ολοκληρώματος ως Δx Σ f(x)
print("Ολοκλήρωμα 1 = {}".format(Result1)) # παρουσιάζουμε το αποτέλεσμα!
```

Ας το επαναλάβουμε για το ολοκλήρωμα $$\int_1^4 \left(\frac{3}{2}\sqrt{x} - \frac{4}{x^2}\right)dx$$

```python
x = np.linspace(1,4,100)                  # διαμερίζουμε ομοιόμορφα το διάστημα [1,4] σε 100 σημεία και τα αποθηκεύουμε στο x!
                                          # Αυτά είναι τα ξ_i μας!
dx = x[1]-x[0]                            # λόγω ομοιόμορφης διαμέρισης, όλα τα διαστήματα Δx_i είναι ίσα με Δx_1 = x[1]-x[0]

func2 = (3/2)*x**(1/2) - 4/(x**2)          # ορίζουμε τη συνάρτηση f(x) και την υπολογίζουμε πάνω στην ομοιόμορφη διαμέριση x
Result2 = dx*np.sum(func2)                 # εκτελούμε το άθροισμα-προσέγγιση του ολοκληρώματος ως Δx Σ f(x)
print("Ολοκλήρωμα 2 = {}".format(Result2)) # παρουσιάζουμε το αποτέλεσμα!
```

Ξέρουμε από πίνακες ολοκληρωμάτων (ή αν εκτελέσουμε την ολοκλήρωση με το "χέρι") ότι η απάντηση είναι $4$, και βλέπουμε ότι όντως έχουμε μια πολύ καλή προσέγγιση!!

Επανάληψη μήτηρ μαθήσεως, άλλη μια φορά! Έστω το $$\int_0^{\pi/6}(1-\cos(3x))\sin(3x)dx$$

```python
x = np.linspace(0,np.pi/6,1000)               # διαμερίζουμε ομοιόμορφα το διάστημα [1,4] σε 1000 σημεία (πιο πολλά τώρα) και 
                                              # τα αποθηκεύουμε στο x! Αυτά είναι τα ξ_i μας!
dx = x[1]-x[0]                                # λόγω ομοιόμορφης διαμέρισης, όλα τα διαστήματα Δx_i είναι ίσα με Δx_1 = x[1]-x[0]

func3 = (1-np.cos(3*x))*np.sin(3*x)           # ορίζουμε τη συνάρτηση f(x) και την υπολογίζουμε πάνω στην ομοιόμορφη διαμέριση x
Result3 = dx*np.sum(func3)                    # εκτελούμε το άθροισμα-προσέγγιση του ολοκληρώματος ως Δx Σ f(x)
print("Ολοκλήρωμα 3 = {}".format(Result3))    # παρουσιάζουμε το αποτέλεσμα!
```

Μπορούμε να υπολογίσουμε με το "χέρι" ότι η απάντηση είναι $1/6$, και βλέπουμε ότι όντως έχουμε μια πολύ καλή προσέγγιση!!

## Ζητούμενο:

Δείξτε ότι τα παρακάτω ολοκληρώματα έχουν τις τιμές που τα συνοδεύουν:

\begin{align}
(α) & \int_0^{\pi}\sin(4x)\cos(5x)dx = -\frac{8}{9} \nonumber \\
\nonumber \\
(β) & \int_0^{2\pi} \frac{dx}{\frac{5}{4} - \cos(x)} = \frac{8\pi}{3} \nonumber \\
\nonumber\\
(γ) & \int_{0^{+}}^1 \frac{\ln(x)}{1+x}dx = -\frac{\pi^2}{12} \nonumber \\
\nonumber\\
(δ) & \int_{0}^{+\infty} \frac{\sin^2(x)}{x^2}dx = \frac{\pi}{2} \nonumber 
\end{align}

Επιβεβαιώστε κάθε φορά ότι το αριθμητικό αποτέλεσμα ισούται με το δοθέν, όπως στο παραπάνω παράδειγμα.

$\textbf{Hint}$ για τα (γ, δ): $\underline{μην}$ ξεκινήσετε τη διαμέρισή σας από το μηδέν ακριβώς αλλά από κάπου ΠΟΛΥ κοντά σε αυτό! Για το (δ), προσεγγίστε το άνω όριο του ολοκληρώματος με ένα μεγάλο αριθμό (π.χ. 1000).

```python
# Γράψτε εδώ τον κώδικά σας

# ------------------------------------------------ 3 ------------------------------------------------ #

import numpy as np                                 # Εισαγωγή βιβλιοθήκης numpy ως np για συντομογραφία.

x = np.linspace(0, np.pi, 1000)                    # Διαμερίζουμε το διάστημα [0, π] σε 1000 σημεία.
dx = x[1] - x[0]                                   # Ισότητα διαστημάτων λόγω ομοιόμορφης διαμέρισης.
funca = np.sin(4 * x) * np.cos(5 * x)              # Ορίζουμε τη συνάρτηση f(x) = sin(4x) * cos(5x).
resa = dx * np.sum(funca)                          # Υπολογίζουμε το ολοκλήρωμα της συνάρτησης.
print("\n⋅ Ολοκλήρωμα (α) =", resa)                # Εκτυπώνουμε το προσεγγιστικό αποτέλεσμα.


x = np.linspace(0, 2 * np.pi, 1000)                # Διαμερίζουμε το διάστημα [0, 2π] σε 1000 σημεία.
dx = x[1] - x[0]                                   # Ισότητα διαστημάτων λόγω ομοιόμορφης διαμέρισης.
funcb = 1 / (5 / 4 - np.cos(x))                    # Ορίζουμε συνάρτηση f(x) = 1 / (5 / 4 - cos(x)).
resb = dx * np.sum(funcb)                          # Υπολογίζουμε το ολοκλήρωμα της συνάρτησης.
print("⋅ Ολοκλήρωμα (β) =", resb)                  # Εκτυπώνουμε το προσεγγιστικό αποτέλεσμα.


x = np.linspace(1e-6, 1, 1000)                     # Διαμερίζουμε διάστημα [0+, 1] πολύ κοντά στο 0.
dx = x[1] - x[0]                                   # Ισότητα διαστημάτων λόγω ομοιόμορφης διαμέρισης.
funcc = np.log(x) / (1 + x)                        # Ορίζουμε τη συνάρτηση f(x) = ln(x) / (1 + x).
resc = dx * np.sum(funcc)                          # Υπολογίζουμε το ολοκλήρωμα της συνάρτησης.
print("⋅ Ολοκλήρωμα (γ) =", resc)                  # Εκτυπώνουμε το προσεγγιστικό αποτέλεσμα.


x = np.linspace(1e-6, 1000, 10000)                 # Διαμερίζουμε διάστημα [0+, 1000] σε 10000 σημεία.
dx = x[1] - x[0]                                   # Ισότητα διαστημάτων λόγω ομοιόμορφης διαμέρισης.
funcd = (np.sin(x) ** 2) / (x ** 2)                # Ορίζουμε τη συνάρτηση f(x) = (sin^2(x)) / x^2).
resd = dx * np.sum(funcd)                          # Υπολογίζουμε το ολοκλήρωμα της συνάρτησης.
print("⋅ Ολοκλήρωμα (δ) =", resd)                  # Εκτυπώνουμε το προσεγγιστικό αποτέλεσμα.

# ------------------------------------------------ 3 ------------------------------------------------ #

# Τέλος κώδικα
```

# Άσκηση Python 4

Σε αυτήν την τελευταία άσκηση θα δείτε πως μπορούμε να παράξουμε όμορφα γραφήματα για τις συναρτήσεις μας!

Η δημιουργία συναρτήσεων γίνεται με τον ίδιο τρόπο που είδατε στην προηγούμενη άσκηση. Διαμερίζουμε το πεδίο ορισμού της και υπολογίζουμε τη συνάρτηση στα διαστήματα αυτά, ακριβώς όπως δείξαμε πριν.

Ας τυπώσουμε τη συνάρτηση 

$$f(x)= \sin(4x)\cos(5x)$$ 

που είδαμε ως ζητούμενο στην προηγούμενη άσκηση. Ας την τυπωσουμε σε ένα διάστημα $[0, 8\pi]$.

```python
x = np.linspace(0, 8*np.pi, 1000)   # 1000 σημεία στο διάστημα [0,8π]

f = np.sin(4*x)*np.cos(5*x)         # ορίζουμε τη συνάρτηση

plt.figure()
plt.plot(x,f)                       # καλούμε τη συνάρτηση plot της matplotlib για την απεικόνιση
plt.xlabel('άξονας x')              # βάζουμε τίτλους στους άξονες και στο γράφημα
plt.title('Συνάρτηση f(x)')         # βάζουμε τίτλους στους άξονες και στο γράφημα
plt.ylabel('f(x)')                  # βάζουμε τίτλους στους άξονες και στο γράφημα
```

Ας το επαναλάβουμε για τη συνάρτηση 

\begin{equation}
f(x) = \frac{\ln(x^2+1)}{x^2+1} \nonumber 
\end{equation}

στο διάστημα $[0, 20]$.

```python
x = np.linspace(0, 20, 1000)       # 1000 σημεία στο διάστημα [0,20]

f = np.log(x**2 + 1)/(x**2 + 1)    # ορίζουμε τη συνάρτηση (η log() είναι η ln() )

plt.figure()
plt.plot(x,f)                       # καλούμε τη συνάρτηση plot της matplotlib για την απεικόνιση
plt.xlabel('άξονας x')              # βάζουμε τίτλους στους άξονες και στο γράφημα
plt.title('Συνάρτηση f(x)')         # βάζουμε τίτλους στους άξονες και στο γράφημα
plt.ylabel('f(x)')                  # βάζουμε τίτλους στους άξονες και στο γράφημα
```

## Ζητούμενο:

Κάντε τις γραφικές παραστάσεις των συναρτήσεων 

\begin{align}
(α) & \: \frac{1}{\frac{5}{4} - \cos(x)} \nonumber \\
\nonumber \\
(β) & \: \frac{\ln(x)}{1+x}\nonumber \\
\nonumber \\
(γ) & \: \frac{(x+2)(x-2)}{x^2}\nonumber 
\end{align}

σε πεδία ορισμού της επιλογής σας.

```python
# Γράψτε εδώ τον κώδικά σας

# -------------------------------------------------------------------- 4 -------------------------------------------------------------------- #

import numpy as np                                                          # Βιβλιοθήκη numpy ως np.
import matplotlib.pyplot as plt                                             # Βιβλιοθήκη matplotlib.pyplot ως plt.

# Συνάρτηση (α): f(x) = 1 / (5 / 4 - cos(x)).
xa = np.linspace(-10, 10, 1000)                                             # Διάστημα [-10, 10] με 1000 σημεία.
fa = 1 / (5 / 4 - np.cos(xa))                                               # Υπολογίζονται οι τιμές για κάθε x.

figa = plt.figure()                                                         # Δημιουργείται το πρώτο γράφημα.
plt.plot(xa, fa)                                                            # Σχεδιάζεται το γράφημα της συνάρτησης.
plt.xlabel("x")                                                             # Ετικέτα του άξονα x.
plt.ylabel("f(x)")                                                          # Ετικέτα του άξονα y.
plt.title("Συνάρτηση (α): f(x) = 1 / (5 / 4 - cos(x))")                     # Τίτλος του γραφήματος.
plt.xlim(-10, 10)                                                           # Όρια του άξονα x.
plt.ylim(-1, 5)                                                             # Όρια του άξονα y.
plt.grid(True)                                                              # Ενεργοποιείται το πλέγμα.
plt.axhline(0, color="black", lw=0.5, ls="--")                              # Σχεδιάζεται ο άξονας x.
plt.axvline(0, color="black", lw=0.5, ls="--")                              # Σχεδιάζεται ο άξονας y.

plt.get_current_fig_manager().set_window_title("⋅ Γραφήματα Συναρτήσεων.")  # Τίτλος του παραθύρου γραφήματος.
plt.show()                                                                  # Εμφανίζεται το γράφημα.

# Συνάρτηση (β): f(x) = ln(x) / (1 + x).
xb = np.linspace(0.1, 10, 1000)                                             # Διάστημα [0.1, 10] με 1000 σημεία.
fb = np.log(xb) / (1 + xb)                                                  # Υπολογίζονται οι τιμές για κάθε x.

figb = plt.figure()                                                         # Δημιουργείται επόμενο γράφημα.
plt.plot(xb, fb)                                                            # Σχεδιάζεται το γράφημα της συνάρτησης.
plt.xlabel("x")                                                             # Ετικέτα του άξονα x.
plt.ylabel("f(x)")                                                          # Ετικέτα του άξονα y.
plt.title("Συνάρτηση (β): f(x) = ln(x) / (1 + x)")                          # Τίτλος του γραφήματος.
plt.xlim(-2, 10)                                                            # Όρια του άξονα x.
plt.ylim(-2, 2)                                                             # Όρια του άξονα y.
plt.grid(True)                                                              # Ενεργοποιείται το πλέγμα.
plt.axhline(0, color="black", lw=0.5, ls="--")                              # Σχεδιάζεται ο άξονας x.
plt.axvline(0, color="black", lw=0.5, ls="--")                              # Σχεδιάζεται ο άξονας y.

plt.get_current_fig_manager().set_window_title("⋅ Γραφήματα Συναρτήσεων.")  # Τίτλος του παραθύρου.
plt.show()                                                                  # Εμφανίζεται το γράφημα.

# Συνάρτηση (γ): f(x) = ((x + 2)(x - 2)) / x ^ 2.
xcleft = np.linspace(-10, -0.1, 500)                                        # Διάστημα [-10, -0.1] με 500 σημεία.
xcright = np.linspace(0.1, 10, 500)                                         # Διάστημα [0.1, 10] με 500 σημεία.
fcleft = ((xcleft + 2) * (xcleft - 2)) / xcleft**2                          # Υπολογίζονται οι τιμές της συνάρτησης.
fcright = ((xcright + 2) * (xcright - 2)) / xcright**2                      # Υπολογίζονται οι τιμές της συνάρτησης.

figc = plt.figure()                                                         # Δημιουργείται επόμενο γράφημα.
plt.plot(xcleft, fcleft, label="x < 0")                                     # Σχεδιάζεται το αριστερό γράφημα.
plt.plot(xcright, fcright, label="x > 0")                                   # Σχεδιάζεται το δεξί γράφημα.
plt.xlabel("x")                                                             # Ετικέτα του άξονα x.
plt.ylabel("f(x)")                                                          # Ετικέτα του άξονα y.
plt.title("Συνάρτηση (γ): f(x) = ((x + 2)(x - 2)) / x ^ 2")                 # Τίτλος του γραφήματος.
plt.xlim(-10, 10)                                                           # Όρια του άξονα x.
plt.ylim(-10, 10)                                                           # Όρια του άξονα y.
plt.grid(True)                                                              # Ενεργοποιείται το πλέγμα.
plt.axhline(0, color="black", lw=0.5, ls="--")                              # Σχεδιάζεται ο άξονας x.
plt.axvline(0, color="black", lw=0.5, ls="--")                              # Σχεδιάζεται ο άξονας y.
plt.legend()                                                                # Εμφανίζεται ο δείκτης του γραφήματος.

plt.get_current_fig_manager().set_window_title("⋅ Γραφήματα Συναρτήσεων.")  # Τίτλος του παραθύρου.
plt.show()                                                                  # Εμφανίζεται το γράφημα.

# -------------------------------------------------------------------- 4 -------------------------------------------------------------------- #

# Τέλος κώδικα
```
