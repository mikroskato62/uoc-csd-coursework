# Άσκηση Python 1

Σε αυτήν την άσκηση θα προσομοιώσετε έναν επιταχυντή σωματιδίων σε Python (το CERN πέφτει λίγο μακριά για να τον βλέπατε στην πραγματικότητα 😃).

Ας υποθέσουμε ότι ο επιταχυντής μας αποτελείται από δυο παράλληλες πλάκες $A, B$, αρκετά κοντά τοποθετημένες μεταξύ τους - απόσταση $d$ - ώστε το ηλεκτρικό πεδίο ανάμεσά τους να θεωρείται ομογενές, όπως στο σχήμα.

![uf](uniform.jpg)

Γνωρίζετε από τη θεωρία σας ότι το ηλεκτρικό πεδίο $\vec{E}$ εγείρει δύναμη 

$$\vec{F} = m\vec{a} = q\vec{E}$$

επάνω σε οποιοδήποτε φορτίο $q$ τοποθετηθεί εντός του. Από την παραπάνω σχέση εύκολα βρίσκουμε ότι

$$\vec{a} = \frac{q\vec{E}}{m}$$

Η δύναμη $\vec{F}$ είναι παράλληλη στις δυναμικές γραμμές του ηλεκτρικού πεδίου, και η φορά της εξαρτάται από το φορτίο που θα τοποθετήσουμε (για την ακρίβεια, εξαρτάται μόνο από το πρόσημο του φορτίου).

Αν τοποθετήσουμε ακίνητο ηλεκτρικό φορτίο $q$ εντός του πεδίου, η κίνηση που αυτό θα κάνει θα είναι ευθύγραμμη ομαλή με σταθερή επιτάχυνση, από τη μια πλάκα προς την άλλη, στον άξονα $x'x$. Θεωρούμε την επίδραση της βαρύτητας αμελητέα.

Από τις εξισώσεις της κινητικής, έχουμε ότι στην $x'x$ κίνησή του, ισχύει:

\begin{align}
x(t) &= x_0 + \frac{1}{2}\frac{qE}{m}t^2 \\
u(t) &= \frac{qE}{m}t
\end{align}

Αν το φορτίο τοποθετηθεί με κάποια αρχική ταχύτητα $u_0$, τότε οι εξισώσεις μας γίνονται

\begin{align}
x(t) &= x_0 + u_{0x}t + \frac{1}{2}\frac{qE}{m}t^2 \\
u(t) &= u_{0x} + \frac{qE}{m}t
\end{align}

## Ζητούμενο 1:

Θεωρήστε ένα ηλεκτρόνιο, με μάζα $m = 9.1 \times 10^{-31}$ kg και ηλεκτρικό φορτίο $q = -1.6 \times 10^{-19}$ C, εντός ηλεκτρικού πεδίου τιμής $E = 0.02$ N/C. Ας γράψουμε μια συνάρτηση που τυπώνει τη θέση και την ταχύτητα που έχει το ηλεκτρόνιο εντός παράλληλων πλακών απόστασης $d=0.01$ m και μήκους $l=0.1$, όταν το αφήσουμε **ακίνητο** σε οποιοδήποτε σημείο. Θα θεωρήσουμε ότι το σημείο αρχής των αξόνων μας, $(0,0)$, βρίσκεται στο κάτω αριστερό μέρος της πλάκας. 

Από τις παραπάνω εξισώσεις εύκολα βρίσκουμε - επιβεβαιώστε το, θεωρώντας θετική φορά κίνησης προς τα δεξιά - ότι ο χρόνος που απαιτείται ώστε το ηλεκτρόνιο να καλύψει την απόσταση από τo σημείο που το αφήνουμε ως την πλάκα που κατευθύνεται είναι $$x(t) = x_0 + \frac{1}{2}\frac{qE}{m}t^2 \Longrightarrow 0 = x_0 - \frac{1}{2}|a|t^2 \Longleftrightarrow t_{max} = \sqrt{\frac{2x_0}{|a|}}$$

με το $|a| = \frac{|q|E}{m}$ στον παρονομαστή να δηλώνει το μέτρο της επιτάχυνσης.

Συμπληρώστε τον κώδικα παρακάτω:

```python
def plot_q_xu(x0, y0):

    m = 9.1 * 10 ** (-31)    # Μάζα ηλεκτρονίου.
    E = 0.02                 # Ηλεκτρικό πεδίο.
    q = -1.6 * 10 ** (-19)   # Φορτίο.
    d = 0.01                 # Απόσταση πλακών.
    l = 0.1                  # Μήκος πλακών.

    assert (x0 > 0) & (x0 < d), "x0 πρέπει να είναι 0 < x0 < d"   # Έλεγχος ότι η είσοδος έχει νόημα με βάση το πρόβλημα.
    assert (y0 > 0) & (y0 < l), "y0 πρέπει να είναι 0 < y0 < l"   # Έλεγχος ότι η είσοδος έχει νόημα με βάση το πρόβλημα.

    a = (q * E / m)                           # Επιτάχυνση.
    tmax = np.sqrt(2 * x0 / abs(a))           # Μέγιστος χρόνος.
    tx = np.linspace(0, tmax, 10000)          # Χρονικός άξονας.
    x = (x0 + 0 * tx + 1 / 2 * a * tx ** 2)   # Θέση.
    u = (0 + a * tx)                          # Ταχύτητα.

    plt.figure(figsize=(9, 6))
    plt.subplot(2, 1, 1)
    plt.plot(tx, x)
    plt.ylabel('x(t)')
    plt.xlabel('Χρόνος (s)')
    plt.subplot(2, 1, 2)
    plt.plot(tx, u)
    plt.ylabel('u(t)')
    plt.xlabel('Χρόνος (s)')
    
#   plt.show()
```

## Ζητούμενο 2:

Βρείτε την εξίσωση τροχιάς ενός **πρωτονίου** (θετικό φορτίο ίσο κατά μέτρο με αυτό του ηλεκτρονίου), χρησιμοποιώντας τις παραπάνω σχέσεις και κατασκευάζοντας μια συνάρτηση $y(x)$. Θεωρήστε ένα πρωτόνιο, με μάζα $m = 9.1 \times 10^{-31}$ kg και ηλεκτρικό φορτίο $q = 1.6 \times 10^{-19}$ C, εντός ηλεκτρικού πεδίου τιμής $E = 0.02$ N/C. Γράψτε μια συνάρτηση που τυπώνει την τροχιά που ακολουθεί το φορτίο εντός παράλληλων πλακών απόστασης $d=0.01$ m και μήκους $l=0.1$ m, όταν του δώσουμε **αρχική ταχύτητα $u_0\vec{i}$** σε οποιοδήποτε σημείο $(x_0, y_0)$. Η συνάρτησή σας, εκτός από το γράφημα, θα πρέπει να απαντά και στο ερώτημα αν το φορτίο θα χτυπήσει στην κάτω πλάκα ή όχι, επιστρέφοντας τη φράση **"Χτύπησε!"** ή **"Δε χτύπησε!"**.

```python
def plot_q_trajectory(x0, y0, u0):

    m = 9.1 * 10 ** (-31)   # Μάζα ηλεκτρονίου.
    E = 0.02                # Ηλεκτρικό πεδίο.
    q = 1.6 * 10 ** (-19)   # Φορτίο.
    d = 0.01                # Απόσταση πλακών.
    l = 0.1                 # Μήκος πλακών.

    a = (q * E / m)   # Επιτάχυνση.
    tmax = (l / u0)   # Μέγιστος χρόνος.

    x = (x0 + u0 * np.linspace(0, tmax, 10000))      # Θέση x με χρονικό άξονα.
    y = (y0 + (a / (2 * u0 ** 2)) * (x - x0) ** 2)   # Θέση y με χρονικό άξονα.

    # Γράφημα:
    plt.figure(figsize=(9, 6))
    plt.plot(x, y)
    plt.axhline(y=d, xmin=0, xmax=len(x), linewidth=1, linestyle='--', color='r')
    plt.axhline(y=0, xmin=0, xmax=len(x), linewidth=1, linestyle='--', color='r')
    plt.axis([0, 0.105, -0.001, 0.0105])
    plt.show()

    if y[-1] > d: # Τι σημαίνει το y[-1] και τι σημαίνει η σύγκριση αυτή στην άσκησή μας? -> Τελευταίο στοιχείο του πίνακα y.
        print("Χτυπάει!")
        return "Χτυπάει!"
    else:
        print("Δεν χτυπάει!")
        return "Δεν χτυπάει!"
```

# Άσκηση Python 2

Στο μάθημα συναντήσαμε ηλεκτρικά πεδία από σημειακά φορτία. Στην άσκηση αυτή θα συζητήσουμε και θα οπτικοποιήσουμε ηλεκτρικά πεδία από σημειακά φορτία. Ήδη γνωρίζετε εποπτικά πως συμπεριφέρεται το ηλεκτρικό πεδίο ενός φορτίου $q$: έχει κατεύθυνση ακτινικά προς τα "έξω" αν $q>0$, ενώ έχει κατεύθυνση ακτινικά προς τα "μέσα" αν $q<0$. Επίσης, είδατε μερικά ηλεκτρικά πεδία από διάφορα σημειακά φορτία τοποθετημένα σε "βολικές" θέσεις. 

Σε αυτήν την άσκηση, θα δούμε πώς μπορούμε να σχεδιάσουμε πεδία (ηλεκτρικά, εν προκειμένω, αλλά στην πραγματικότητα οποιαδήποτε) με κάποιες πολύ βοηθητικές συναρτήσεις της matplotlib.

Ας υποθέσουμε ότι έχουμε ένα θετικό φορτίο $q$ τοποθετημένο σε ένα **οποιοδήποτε** σημείο του επιπέδου, σε απόσταση $r_0$ από τη συμβολή των αξόνων του συστήματος συντεταγμένων μας. Έστω ότι θέλουμε να βρούμε το ηλεκτρικό πεδίο σε $x-$ και $y-$ συνιστώσες σε ένα **οποιοδήποτε** άλλο σημείο του επιπέδου. Δείτε το παρακάτω σχήμα.

![elfield](elfield.png)

Το ηλεκτρικό πεδίο στο τυχαίο σημείο $(x,y)$ θα έχει συνιστώσες όπως στο σχήμα. Ας τις υπολογίσουμε!

Η τιμή του πεδίου θα είναι

$$E = k_e \frac{q}{r^2} = k_e \frac{q}{(x-x_0)^2 + (y-y_0)^2}$$

με $r$ την απόσταση του φορτίου $q$ από το σημείο με συντεταγμένες $(x,y)$. Οι συνιστώσες του ηλεκτρικού πεδίου θα είναι

$$E_x = E\cos(\theta)$$ 

και 

$$E_y = E \sin(\theta)$$

με $\theta$ τη γωνία που φαίνεται στο σχήμα και σχηματίζουν το διάνυσμα $\vec{E}_x$ με το διάνυσμα $\vec{E}$. Από την τριγωνομετρία του σχήματος κοντά στο φορτίο $q$ φαίνεται ότι 

\begin{align}
\cos(\theta) &= \frac{x-x_0}{r} = \frac{x-x_0}{\sqrt{(x-x_0)^2 + (y-y_0)^2}} \\
\sin(\theta) &= \frac{y-y_0}{r} = \frac{y-y_0}{\sqrt{(x-x_0)^2 + (y-y_0)^2}}
\end{align}

και άρα

\begin{align}
E_x &= k_e \frac{q}{(x-x_0)^2 + (y-y_0)^2}\frac{x-x_0}{\sqrt{(x-x_0)^2 + (y-y_0)^2}} = k_e \frac{q(x-x_0)}{\left((x-x_0)^2+(y-y_0)^2\right)^{\frac{3}{2}}}\\
E_y &= k_e \frac{q}{(x-x_0)^2 + (y-y_0)^2}\frac{y-y_0}{\sqrt{(x-x_0)^2 + (y-y_0)^2}} = k_e \frac{q(y-y_0)}{\left((x-x_0)^2+ (y-y_0)^2\right)^{\frac{3}{2}}}
\end{align}

Ουφ! Βρήκαμε επιτέλους το ηλεκτρικό πεδίο σε συνιστώσες για *οποιοδήποτε* σημείο του επιπέδου, αν μια πηγή φορτίου $q$ βρίσκεται (όχι στη συμβολή των αξόνων, όπως βολικά υποθέταμε στη θεωρία) *οπουδήποτε* στο επίπεδο!

## Ζητούμενο 1:
Γράψτε συνάρτηση η οποία υπολογίζει τις συνιστώσες του ηλεκτρικού πεδίου, δεδομένου μιας πηγής φορτίου $q$, της θέσης της $(x_0, y_0)$, και το σημείο του επιπέδου $(x,y)$ που θέλουμε να υπολογίσουμε το πεδίο. Η συνάρτηση θα επιστρέφει τις τιμές των δυο συνιστωσών.

```python
def E(q, x0, y0, x, y):
    
    ke = 9 * 10 ** 9
    
    Ex = ((ke * q * (x - x0)) / (((x - x0) ** 2 + (y - y0) ** 2) ** (3 / 2)))
    Ey = ((ke * q * (y - y0)) / (((x - x0) ** 2 + (y - y0) ** 2) ** (3 / 2)))
    
    return Ex, Ey
    
# Τέλος κώδικα!
```

Ας τη δοκιμάσουμε για δυο περιπτώσεις, σχετικά απλές:

```python
Ex, Ey = E(2, 1, 1, 2, 2)   # Πρέπει να πάρετε (6363961030.678927, 6363961030.678927).
print("Το ηλεκτρικό πεδίο μιας πηγής φορτίου {} C που βρίσκεται στη θέση ({},{}) στο σημείο ({},{}) είναι Ex={}, Ey={}".format(2, 1, 1, 2, 2, Ex, Ey))
Ex, Ey = E(-1, 0, 0, 0, 1)   # Πρέπει να πάρετε (0.0, -9000000000.0).
print("Το ηλεκτρικό πεδίο μιας πηγής φορτίου {} C που βρίσκεται στη θέση ({},{}) στο σημείο ({},{}) είναι Ex={}, Ey={}".format(-1, 0, 0, 0, 1, Ex, Ey))
```

Μόλις παραπάνω δοκιμάσαμε ένα φορτίο (θετικό ή αρνητικό) σε δυο σημεία και υπολογίσαμε το ηλεκτρικό πεδίο σε δυο άλλα σημεία. Όμως για την απεικόνιση του ηλεκτρικού πεδίου θέλουμε πολλά σημεία, όσο το δυνατόν περισσότερα. Ας φτιάξουμε ένα "πλέγμα" του $xy-$επιπέδου, επιλέγοντας αρκετά "πυκνά" διάφορες συντεταγμένες αυτού του επιπέδου. Για ευκολία, ας θεωρήσουμε ότι το "πλέγμα" μας θα βρίσκεται εντός του υπο-επιπέδου $[-2,2] \times [-2,2]$.

```python
nx, ny = 64, 64                # Θα ζητήσουμε 64 σημεία x 64 σημεία στο υποεπίπεδο [-2,2]x[-2,2].
x = np.linspace(-2, 2, nx)     # Φτιάχνουμε τον άξονα x'x του υποεπιπέδου.
y = np.linspace(-2, 2, ny)     # Φτιαχνουμε τον άξονα y'y του υποεπιπέδου.
X, Y = np.meshgrid(x, y)       # Καλούμε τη συνάρτηση meshgrid της numpy που μας φτιάχνει το "πλέγμα" 64x64 σημείων εντ΄ός του [-2,2]x[-2,2].
```

Αυτό που φτιάξαμε στην ουσία μοιάζει με το παρακάτω σχήμα, με τη μόνη διαφορά ότι υπάρχουν πολλές περισσότερες "τελίτσες" (σημεία) στο $[-2,2]\times [-2,2]$.

![grid](grid.png)

Ας πάμε τώρα να υπολογίσουμε το ηλεκτρικό πεδίο και να το απεικονίσουμε μέσω των δυναμικών του γραμμών! Πρώτα όμως να τοποθετήσουμε ένα ή περισσότερα φορτία στο χώρο!

```python
Nq = 3                                   # Πόσα φορτία θέλουμε ? Έστω 3.
charges = []                             # Θα αποθηκεύουμε τιμή φορτίου (1 ή -1 C) και συντεταγμένες στο (x,y) επίπεδο.
for i in range(Nq):
    q = i % 2 * 2 - 1                    # Φορτίο 1 ή -1 C, εναλλάξ.
    charges.append((q, 0, float(i/2)))   # Θα τα βάλουμε πάνω στην ευθεία x=0, στα σημεία y=0, y=1/2, y=1.
    print("Ένα φορτίο q={} C βρίσκεται στη θέση x0={}, y0={}".format(q, 0, float(i/2)))
```

Ας υπολογίσουμε τώρα το ηλεκτρικό πεδίο σε όλο το πλέγμα $64$ x $64$ (που είπαμε είναι σημεία ομοιόμορφα κατανεμημένα μέσα στο $[-2,2]\times[-2,2]$), για τις παραπάνω πηγές φορτίου!

```python
Ex, Ey = np.zeros((ny, nx)), np.zeros((ny, nx))             # Δεσμεύουμε χώρο στη μνήμη για να κρατήσουμε τις τιμές του πεδίου.
for charge in charges:                                      # Για κάθε φορτίο που ορίσαμε.
    ex, ey = E(charge[0], charge[1], charge[2], x=X, y=Y)   # Υπολογίζουμε το ηλεκτρικό πεδίο σε όλο το "πλέγμα".
    Ex += ex                                                # Προσθέτουμε τις x-συνιστώσες μαζί.
    Ey += ey                                                # Προσθέτουμε τις y-συνιστώσες μαζί.
```

Σε αυτό το σημείο τι έχουμε καταφέρει? Έχουμε υπολογίσει το ηλεκτρικό πεδίο ανά συνιστώσες για διάφορα σημεία του $[-2,2]\times[-2,2]$, και συγκεκριμένα σε ένα πλέγμα $64$ x $64$, δηλ. σε $4096$ σημεία ομοιόμορφα κατανεμημένα μέσα στο $[-2,2]\times[-2,2]$!!

Ο παρακάτω κώδικας σχεδιάζει τις δυναμικές γραμμές του ηλεκτρικού πεδίου που σχηματίζεται από τα τρια φορτία και είναι εκτός πλαισίου μαθήματος. Απλά τρέξτε τον, κατανοήστε πως λειτουργεί (πάνω-κάτω) και χρησιμοποιήστε τον παρακάτω όπου χρειαστεί.

# Άσκηση Python 3

Σε αυτήν την άσκηση θα δούμε έναν τρόπο να λύνουμε πολύπλοκα κυκλώματα με χρήση γραμμικής άλγεβρας. 

Αν παρακολουθείτε το μάθημα ΗΥ119-Γραμμική Άλγεβρα θα έχετε ήδη διδαχθεί μεθόδους εύρεσης λύσης σε $N$ x $N$ συστήματα εξισώσεων. Αν πάλι δεν το παρακολουθείτε, δεν πειράζει. 

Δείτε το παρακάτω κύκλωμα.

![circlab0](circlab.png)

Θέλουμε να βρούμε τα ρεύματα του κυκλώματος αυτού. Ξέρετε ότι ένας καλός τρόπος είναι να χρησιμοποιήσουμε τους κανόνες του Kirchhoff έτσι ώστε να καταλήξουμε σε ένα σύστημα $N$ εξισώσεων με $N$ αγνώστους. Συνήθως θα βάζαμε περισσότερα ρεύματα από όσα φαίνονται στο σχήμα για να καλύψουμε όλους τους αντιστάτες. Δείτε για παράδειγμα τον κόμβο $d$. Στον κόμβο αυτό, μπαίνει το ρεύμα $I_1$ και βγαίνει το ρεύμα $I_2$. Το ρεύμα που διαρρέει τον αντιστάτη των $3$ Ω θα είναι $I_1-I_2$. Παρόμοιες υποθέσεις μπορούμε να κάνουμε και σε άλλους αντιστάτες που το ρεύμα τους δεν έχει σημειωθεί στο σχήμα - αρκεί φυσικά σε κάθε κόμβο να ισχύει ο 1ος κανόνας του Kirchhoff: $I_{in} = I_{out}$, δηλ. προσέξτε να μη δημιουργήσετε κόμβο που όλα τα ρεύματα φεύγουν από αυτόν ή όλα τα ρεύματα μπαίνουν σε αυτόν! 😃 Αν το κάνετε σωστά (και δεν είναι εύκολο), θα πάρετε το κύκλωμα του παρακάτω σχήματος.

![circlab](circlab3.png)

Ας εφαρμόσουμε τον 2ο κανόνα του Kirchhoff λοιπόν:

1. Κανόνας βρόχου (αριστερά, πάνω): $-2I_1 - 3(I_1-I_2) + 5 = 0 \Longleftrightarrow 5I_1 - 3I_2 = 5$
2. Κανόνας βρόχου (δεξιά, πάνω): $4 + 3(I_1 - I_2) - 6(I_2 - I_3) = 0 \Longleftrightarrow 3I_1 - 9I_2 + 6I_3 = -4$
3. Κανόνας βρόχου (κάτω): $6 - 5 - 6(I_2 - I_3) - 4I_3 = 0 \Longleftrightarrow  10I_3 - 6I_2 = 1$

Έχουμε λοιπόν το σύστημα $3$ x $3$:

\begin{align}
5I_1 - 3I_2 + 0I_3= 5 \\
3I_1 - 9I_2 + 6I_3 = -4 \\
0I_1 - 6I_2 + 10I_3 = 1
\end{align}

Τους μηδενικούς συντελεστές σε κάποια ρεύματα τα βάλαμε για λόγους που θα δείτε παρακάτω.

Λύνουμε το σύστημα: $$I_3 = \frac{1 + 6I_2}{10}$$ και $$3I_1 - 9I_2 + 6\frac{1 + 6I_2}{10} = 4 \Longleftrightarrow I_1 = \frac{27I_2 - 23}{15}$$ και άρα:

$$5\frac{27I_2 - 23}{15} - 3I_2 = 5 \Longleftrightarrow \frac{18}{3}I_2 = \frac{38}{3} \Longleftrightarrow I_2 = \frac{19}{9} \approx 2.111 A$$

Τέλος, $$I_1 = -\frac{23}{15} + \frac{27}{15}\frac{19}{9} \Longleftrightarrow I_1 = \frac{34}{15} \approx 2.2667 A$$

και

$$10I_3 - 6\frac{19}{9} = 1 \Longleftrightarrow I_3 = \frac{41}{30} \approx 1.3667 A$$

Είδατε ότι θέλει πολλή προσοχή στην επίλυση του συστήματος. Πώς μπορούν να μας βοηθήσουν οι υπολογιστές στην επίλυση τέτοιων δύσκολων συστημάτων? Στο κάτω-κάτω, σε αυτό είναι καλοί οι υπολογιστές: στις πράξεις!

Κατασκευάζουμε έναν **πίνακα** (που δεν είναι τίποτε άλλο από μια διάταξη αριθμών σε γραμμές και στήλες, όπως θα δείτε παρακάτω) $A$ και δυο διανύσματα $I$ και $b$ κοιτάζοντας το σύστημα $3$ x $3$ παραπάνω, το οποίο επαναλαμβάνουμε εδώ για ευκολία:

\begin{align}
5I_1 - 3I_2 + 0I_3= 5 \\
3I_1 - 9I_2 + 6I_3 = -4 \\
0I_1 - 6I_2 + 10I_3 = 1
\end{align}

ως εξής. Ο παρακάτω είναι ο λεγόμενος πίνακας $A$.

$$ A = \left(\begin{array}{ccc}
5 & -3 & 0 \\
3 & -9 & 6 \\
0 & -6 & 10
\end{array}\right)$$

Ενώ παρακάτω βλέπετε τα διανύσματα $I$ και $b$.

$$I = \left(\begin{array}{ccc}
I_1 \\
I_2 \\
I_3 
\end{array}
\right), \: \: \: b = \left(\begin{array}{ccc}
5 \\
-4 \\
1 
\end{array}
\right)$$

Ο πίνακας $A$ περιέχει τους συντελεστές των ρευμάτων κάθε εξίσωσης σε μια γραμμή του, σε αύξουσα σειρά (πρώτα ο συντελεστής του $I_1$, μετά του $I_2$ και τέλος του $I_3$). Το διάνυσμα με τα άγνωστα ρεύματα πρέπει κι αυτό να τα περιέχει σε αύξουσα σειρά. Ο σταθερός όρος κάθε εξίσωσης μπαίνει στην αντίστοιχη γραμμή του διανύσματος $b$.

Μετά την εύκολη κατασκευή των $A, I, b$, πολύ απλά πρέπει να λύσουμε το σύστημα 

$$AI = b \Longleftrightarrow \left(\begin{array}{ccc}
5 & -3 & 0 \\
3 & -9 & 6 \\
0 & -6 & 10
\end{array}\right)\left(\begin{array}{ccc}
I_1 \\
I_2 \\
I_3 
\end{array}
\right) = \left(\begin{array}{ccc}
5 \\
-4 \\
1 
\end{array}
\right)$$

που πρακτικά είναι μια διαφορετική γραφή του συστήματος των εξισώσεων (1, 2, 3) που είδατε παραπάνω.

Θα δείξουμε πως μπορούμε να λύσουμε το σύστημα με τη βοήθεια της Python.

```python
A = np.array([[5, -3, 0], [3, -9, 6], [0, -6, 10]])   # Κατασκευή πίνακα Α.
b = np.array([5, -4, 1])                              # Κατασκευή διανύσματος b.
I = np.linalg.solve(A, b)                             # Κατασκευή λύσης με τη βοήθεια της solve από τη βιβλιοθήκη linalg.
```

```python
print("I_1 = {}, I_2 = {}, I3 = {}".format(I[0], I[1], I[2]))   # [2.26667 2.1111 1.36667], ίδιες λύσεις με το "χαρτί" μας.
```
